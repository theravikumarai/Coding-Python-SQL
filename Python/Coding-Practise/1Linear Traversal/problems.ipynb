````python
def second_largest(nums):
    if len(nums) < 2:
        return None

    highest = nums[0]
    second = None

    for n in nums:

        if n > highest:
            second = highest
            highest = n

        elif n != highest and (second is None or n > second):
            second = n

    return second

num = [10, 10, 10]
print(second_largest(num))

# Complexity
# Time
# One traversal.
# O(n)

# Space
# Only:
# highest
# second
# Two variables.
# O(1)
````
