    Interview Hint

    Before writing code, think:

    What data structure should store the frequencies?
    What should happen when a character appears for the first time?
    What should happen when it appears again?

```python
# using nested loop TC -- O(N**2)
def character_frequency(s):
    frequency = {}

    for ch in s:
        count = 0
        for freq in s:
            if ch == freq:
                count += 1
        frequency[ch]=count
    return frequency

text = 'banana'
print(character_frequency(text))
                
```

```python
# Using Hashing
def character_frequency(s):
    frequency = {}

    for ch in s:
        if ch in frequency:
            frequency[ch] += 1
        else:
            frequency[ch] = 1

    return frequency

text = 'banana'
print(character_frequency(text))

# Time Complexity
# One traversal.
# O(n)

# Space Complexity
# Dictionary stores unique characters.
# Worst case:
# "abcdef"

# Dictionary:

# {
# 'a':1,
# 'b':1,
# 'c':1,
# 'd':1,
# 'e':1,
# 'f':1
# }

# Space:

# O(n)
```

```python
# Even More Pythonic

# Once you're comfortable with hashing, you'll often see:

def character_frequency(s):
    frequency = {}

    for ch in s:
        frequency[ch] = frequency.get(ch, 0) + 1

    return frequency

# We'll start using .get() from the next few questions because it's very common in production code.
```
