```python
def word_frequency(sentence):
    frequency = {}
    words = sentence.split()
  
    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1

    return frequency

text = "python is easy python"

print(word_frequency(text))


# Dictionary updates:

# {}
# ↓
# {'python':1}
# ↓
# {'python':1,'is':1}
# ↓
# {'python':1,'is':1,'easy':1}
# ↓
# {'python':2,'is':1,'easy':1}

# Correct.

# Time Complexity

# Suppose there are n words.

# split() → O(n) (processes the string once)
# Traversing the words → O(n)

# Overall:

# O(n)
# Space Complexity

# Worst case:

# Every word is unique.

# {
# word1:1,
# word2:1,
# word3:1,
# ...
# }
# O(n)
```

```python
# Pythonic Version
def word_frequency(sentence):
    frequency = {}

    for word in sentence.split():
        frequency[word] = frequency.get(word, 0) + 1

    return frequency

# Notice that we don't even need a separate words variable unless we want one for readability.
```
