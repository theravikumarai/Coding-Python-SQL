Interview Hint

Think about this:

Suppose:

nums = [1,2,2,3,4,4]

As you traverse:

First 2

Have I seen it before?

No

→ Keep it

Second 2

Have I seen it before?

Yes

→ Ignore it

So what data structure remembers what you've already seen?

Exactly:

seen = set()

```python
# better pythonic way
def remove_duplicates(nums):
    result = []
    seen = set()

    for n in nums:
        if n not in seen:
            seen.add(n)
            result.append(n)

    return result

num = [1,2,3,3]
print(remove_duplicates(num))
```

Time Complexity

For every element:

if n in seen

Average lookup in a set:

O(1)

Adding to a set:

O(1)

Appending to a list:

O(1)

For n elements:

Time = O(n)
