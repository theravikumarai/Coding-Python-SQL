```python
# Optimized Version:

def find_duplicates(nums):
    seen = set()
    duplicate_seen = set()
    result = []

    for n in nums:
        if n in seen:
            if n not in duplicate_seen:
                duplicate_seen.add(n)
                result.append(n)

        else:
            seen.add(n)

    return result
```
