```python
# Interview Prefered Way:

def first_non_repeating(s):
    frequency = {}
    for ch in s:
        if ch in frequency:
            frequency[ch] += 1
        else:
            frequency[ch] = 1

    for ch in s:
        if frequency[ch] == 1:
            return ch

    return None


text = "aabbceccc"
print(first_non_repeating(text))


# Interview solution:

# Build dictionary → O(n)
# Traverse string → O(n)

# Overall:
# O(n)
# Both are asymptotically the same.

# Space Complexity

# Dictionary:
# O(k)

# Worst case:
# O(n)
```
