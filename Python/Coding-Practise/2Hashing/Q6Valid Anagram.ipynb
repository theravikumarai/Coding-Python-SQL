```python
# Cleane Version

def is_anagram(s, t):

    if len(s) != len(t):
        return False

    frequency1 = {}
    frequency2 = {}

    for ch in s:
        frequency1[ch] = frequency1.get(ch, 0) + 1

    for ch in t:
        frequency2[ch] = frequency2.get(ch, 0) + 1

    return frequency1 == frequency2
```

frequency1.get(ch, 0): This translates to: "Look up the character ch in the dictionary. If it exists, give me its current count. If it does not exist yet, don't crash—just give me a default value of 0."

+ 1: Adds 1 to whatever value was fetched (either the existing count or the fallback 0).
