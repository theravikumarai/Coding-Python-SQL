```python
def common_elemets(nums1, nums2):
    seen = set(nums1)
    added = set()
    result = []

    for n in nums2:
        if n in seen and n not in added:
            result.append(n)
            added.add(n)
        else:
            pass
    return result

nums1 = [3,1,1]
nums2 = [1,2,1]

print(common_elemets(nums1,nums2))

# Optimized Solution
# seen = set(nums1)
# added = set()

# Both lookups:

# O(1)

# Overall:

# O(n + m)
```
