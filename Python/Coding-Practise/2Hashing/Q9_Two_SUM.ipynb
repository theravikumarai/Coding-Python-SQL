```python
def two_sum(nums, target):
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i]+nums[j] == target:
                return i,j

nums = [2,3,6,8,9]
target = 14
print(two_sum(nums,target))

# TC - O(N**2)
# SC - O(1)
```

```python
def two_sum(nums, target):

    num_dict = {}

    for i in range(len(nums)):
        num_dict[nums[i]] = i

    for j in range(len(nums)):
        req = target - nums[j]

        if req in num_dict:
            return j, num_dict[req]

num = [1,3,6,8]
target = 9
two_sum(num,target)
```

```python
def two_sum(nums, target):
    lookup = {}

    for index, num in enumerate(nums):
        complement = target - num

        if complement in lookup:
            return lookup[complement], index

        lookup[num] = index

# Complexity
# Time

# One traversal.

# O(n)
# Space

# Dictionary stores every element.

# O(n)
```
