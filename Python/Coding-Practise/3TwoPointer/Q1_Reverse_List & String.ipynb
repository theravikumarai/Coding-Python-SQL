```python
# Reverse a string

def reverse_array(nums):
    left = 0
    right = len(nums) - 1
    temp = 0
    while left < right:
        temp = nums[left]
        nums[left] = nums[right]
        nums[right] = temp

        left += 1
        right -= 1
    return nums

nums = [1,2,3,4,5]

print(reverse_array(nums))
```

```python
def reverse_array(nums):
    left = 0
    right = len(nums) - 1
    temp = 0

    while left < right:
        temp = nums[left]
        nums[left] = nums[right]
        nums[right] = temp

        left += 1
        right -= 1   # right pointer moving to left side

    return nums


# Complexity
# Time

# Each element is swapped once.

# O(n)
# Space

# Extra variables:

# left
# right
# temp
# O(1)
```

```python
# Pythonic Version, tuple unpacking

def reverse_array(nums):
    left = 0
    right = len(nums) - 1

    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1

    return nums
```

```python
# Reverse a string
def reverse_string(text):
    text = list(text)
    left = 0
    right = len(text) - 1

    while left < right:
        text[left], text[right] = text[right], text[left]
        left += 1
        right -= 1

    return "".join(text)

text = "Ravi"
print(reverse_string(text))
```
