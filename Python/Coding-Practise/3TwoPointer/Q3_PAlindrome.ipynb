```python
def is_palindrom(s):
    s = list(s)
    old = s.copy()
    left = 0
    right = len(s) - 1
    while left < right:
        s[left], s[right] = s[right], s[left] 

        left += 1
        right -= 1

    return (s==old)

s = ""
print(is_palindrom(s))
```

```python
# Optimized solution
def is_palindrom(s):

    left = 0
    right = len(s) - 1
    while left < right:

        if s[left] != s[right]:
            return False

        left += 1
        right -= 1

    return True

s = "madam"
print(is_palindrom(s))
```
