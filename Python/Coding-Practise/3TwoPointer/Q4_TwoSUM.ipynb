```python
def two_sum_sorted(nums, target):

    left = 0
    right = len(nums) - 1
    

    while left < right:
        current_sum = nums[left] + nums[right]
        if current_sum == target:
            return left, right

        if current_sum < target:
            left += 1
            

        elif current_sum > target:
            right -= 1
    return None         # as only one solution

nums = [1,2,3,4,5]
target = 7

print(two_sum_sorted(nums, target))
        
```
