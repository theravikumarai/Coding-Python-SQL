```python
def remove_duplicates(nums):

    if not nums:
        return 0

    slow = 0

    for fast in range(1, len(nums)):

        if nums[fast] != nums[slow]:
            slow += 1
            nums[slow] = nums[fast]

    return slow, nums[:slow]

nums = []
print(remove_duplicates(nums))
```

```python
# interview Version
def remove_duplicates(nums):

    if not nums:
        return 0

    slow = 0

    for fast in range(1, len(nums)):

        if nums[fast] != nums[slow]:
            slow += 1
            nums[slow] = nums[fast]

    return slow + 1

nums = [1,2,3,3,4]
print(remove_duplicates(nums))
```
