```python
def move_zeros(nums):

    slow = 0

    for fast in range(len(nums)):

        # if nums[fast] == 0:
        #     continue

        if nums[fast] != 0:
            nums[slow], nums[fast] = nums[fast], nums[slow]
            slow += 1

    return nums





nums = [0, 1, 0, 2 ,3 ,4]
print(move_zeros(nums))
```
