```python
def sorted_arrays(nums):
    result = [0] * len(nums)
    left = 0
    right = len(nums) - 1
    result_idx = len(nums) - 1

    while left <= right:

        if nums[left]**2 >= nums[right]**2:
            result[result_idx] = nums[left]**2
            left += 1
            result_idx -= 1

        else:
            result[result_idx] = nums[right]**2
            result_idx -= 1
            right -= 1

    return result

nums = [-7,2]
print(sorted_arrays(nums))




```
