```python
def max_area(height):

    left = 0
    right = len(height) - 1
    highest_area = 0

    

    while left < right:
        width = right - left
        Height = min(height[left], height[right])
        current_area = width * Height

        # 3. Track the largest area found so far
        if current_area > highest_area:
            highest_area = current_area
        
        if height[left] <= height[right]:
            left += 1
        else:
            right -= 1


    return highest_area

height = [1,8,6,2,5,4,8,3]
print(max_area(height))

```
