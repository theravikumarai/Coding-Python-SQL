```python
def first_negative_in_window(nums, k):

    left = 0
    negatives = []      # Stores only negative numbers
    result = []

    for right in range(len(nums)):

        # Step 1: Add new negative number
        if nums[right] < 0:
            negatives.append(nums[right])

        # Step 2: Window reached size k
        if right - left + 1 == k:

            # Step 3: First negative
            if negatives:
                result.append(negatives[0])
            else:
                result.append(0)

            # Step 4: Remove outgoing negative
            if negatives and nums[left] == negatives[0]:
                negatives.pop(0)

            # Step 5: Slide the window
            left += 1

    return result
```

```python
# Optimal Solution

from collections import deque

def first_negative_in_window(nums, k):
    left = 0
    negatives = deque()
    result = []

    for right in range(len(nums)):

        if nums[right] < 0:
            negatives.append(nums[right])

        if right - left + 1 == k:

            if negatives:
                result.append(negatives[0])
            else:
                result.append(0)

            if negatives and nums[left] == negatives[0]:
                negatives.popleft()

            left += 1

    return result
```
