```python
def max_vowels(s,k):
    left = 0
    vowel_count = 0
    max_vowel = 0
    vowel = "aeiou"

    for right in range(len(s)):

        if s[right] in vowel:
            vowel_count += 1

        if right - left + 1 == k:

            max_vowel = max(max_vowel, vowel_count)

            if s[left] in vowel:
                vowel_count -= 1

            left += 1

    return max_vowel

s = "My name is Ravi"
k = 3
print(max_vowels(s,k))
```

The right pointer visits each character exactly once. The left pointer also moves only forward and never revisits a character. Each character enters the window once and leaves the window once, and all operations inside the loop are constant time. 

    Therefore, the overall time complexity is O(n), and the space complexity is O(1).
