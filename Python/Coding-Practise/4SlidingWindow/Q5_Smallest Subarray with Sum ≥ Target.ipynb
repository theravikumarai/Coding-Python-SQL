```python
def min_subarray_len(target, nums):

    left = 0
    window_sum = 0
    min_length = float("inf")

    for right in range(len(nums)):

        # Step 1: Expand the window
        window_sum += nums[right]

        # Step 2: Shrink the window as long as it satisfies the condition
        while window_sum >= target:

            # Step 3: Update the minimum length
            min_length = min(min_length, right - left + 1)

            # Step 4: Remove the leftmost element
            window_sum -= nums[left]

            # Step 5: Shrink the window
            left += 1

    if min_length == float("inf"):
        return 0

    return min_length


target = 7
nums = [2,3,1,2,4,3]

print(min_subarray_len(target, nums))
```
