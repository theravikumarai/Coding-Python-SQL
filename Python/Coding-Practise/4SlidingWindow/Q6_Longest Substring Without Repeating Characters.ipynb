````python
def length_of_longest_substring(s):
    left = 0
    seen = set()
    max_length = 0

    for right in range(len(s)):

        while s[right] in seen:
            seen.remove(s[left])
            left += 1

        seen.add(s[right])

        max_length = max(max_length, right - left + 1)

    return max_length

s = "abcabcd"
print(length_of_longest_substring(s))
````

# Why `while` instead of `if`?

This is the most important interview question.

Consider:

```python
s = "abba"
```

When `right` reaches the second `'b'`:

Current window:

```text
abb
```

If you used:

```python
if s[right] in seen:
```

you would remove only one character (`'a'`).

Window would become:

```text
bb
```

There is still a duplicate `'b'`.

The window is still invalid.

Using `while` keeps removing characters until the duplicate disappears.

That's why:

```python
while s[right] in seen:
```

is necessary.

# Time Complexity

Many beginners think:

> There is a `while` inside a `for`, so is it `O(n²)`?

No.

### Right Pointer

Moves:

```text
0 → 1 → 2 → ... → n-1
```

Exactly `n` times.

### Left Pointer

Also only moves forward.

Example:

```text
0 → 1 → 2 → 3
```

It never goes backward.

So each character:

* Enters the window once.
* Leaves the window once.

Total operations:

```text
2n
```

Big-O ignores constants:

```text
O(n)
```

---

# Space Complexity

The `set` stores unique characters in the current window.

Worst case:

```python
s = "abcdefg"
```

The set contains:

```text
{a,b,c,d,e,f,g}
```

If the string has `n` unique characters:

```text
O(n)
```

If the alphabet is fixed (e.g., only lowercase English letters), the set can hold at most 26 characters, which is effectively `O(1)`. In general interview analysis, we state:

```text
Space Complexity = O(min(n, alphabet_size))
```
