```python
def character_replacement(s, k):

    left = 0
    frequency = {}
    max_frequency = 0
    max_length = 0

    for right in range(len(s)):

        # Add current character
        ch = s[right]

        if ch in frequency:
            frequency[ch] += 1
        else:
            frequency[ch] = 1

        # Update highest frequency seen in current window
        max_frequency = max(max_frequency, frequency[ch])

        # If window is invalid, shrink it
        while (right - left + 1) - max_frequency > k:

            frequency[s[left]] -= 1
            left += 1

        # Update answer
        max_length = max(max_length, right - left + 1)

    return max_length
```
