```python
def check_inclusion(s1, s2):

    if len(s1) > len(s2):
        return False

    pattern = {}
    window = {}

    # Frequency of s1
    for ch in s1:
        if ch in pattern:
            pattern[ch] += 1
        else:
            pattern[ch] = 1

    left = 0

    for right in range(len(s2)):

        # Add current character
        ch = s2[right]

        if ch in window:
            window[ch] += 1
        else:
            window[ch] = 1

        # Window reached required size
        if right - left + 1 == len(s1):

            # Compare frequencies
            if window == pattern:
                return True

            # Remove left character
            window[s2[left]] -= 1

            if window[s2[left]] == 0:
                del window[s2[left]]

            left += 1

    return False
```
