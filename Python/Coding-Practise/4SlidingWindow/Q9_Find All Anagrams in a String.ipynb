```python
def find_anagrams(s, p):

    # If pattern is longer than string, impossible
    if len(p) > len(s):
        return []

    pattern = {}
    window = {}

    # Build frequency map of pattern
    for ch in p:
        if ch in pattern:
            pattern[ch] += 1
        else:
            pattern[ch] = 1

    left = 0
    result = []

    for right in range(len(s)):

        # Add current character
        ch = s[right]

        if ch in window:
            window[ch] += 1
        else:
            window[ch] = 1

        # Window becomes valid
        if right - left + 1 == len(p):

            # Compare frequencies
            if window == pattern:
                result.append(left)

            # Remove left character
            window[s[left]] -= 1

            if window[s[left]] == 0:
                del window[s[left]]

            left += 1

    return result
```
