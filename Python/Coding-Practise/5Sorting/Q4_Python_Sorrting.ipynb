```python
def sort_students(students):

    return sorted(students, key = lambda x: x[1], reverse=True)

students = [
    ("Ravi", 85),
    ("Amit", 92),
    ("John", 78),
    ("Neha", 95)
]

print(sort_students(students))
```

```python
def sort_by_value(marks):
    return sorted(marks.items(), key= lambda x:x[1])


marks = {
    "Ravi": 85,
    "Amit": 92,
    "John": 78,
    "Neha": 95
}
print(sort_by_value(marks))

```

```python
def sort_by_length(words):

    return sorted(words, key = lambda x: len(x) )

words = ["python", "java", "c", "javascript", "go"]
print(sort_by_length(words))
```

```python
def k_largest(nums,k):

    return sorted(nums,reverse= True)[:k]


nums = [3, 2, 1, 5, 6, 4]
k = 2
print(k_largest(nums,k))
```
