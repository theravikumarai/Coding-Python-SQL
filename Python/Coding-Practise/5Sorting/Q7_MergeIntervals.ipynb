```python
def merge_intervals(intervals):

    # Step 1: Sort intervals by starting value
    intervals.sort(key=lambda x: x[0])

    # Step 2: Store the first interval
    result = [intervals[0]]

    # Step 3: Process remaining intervals
    for interval in intervals[1:]:

        # Check if current interval overlaps
        # with the last interval in result
        if interval[0] <= result[-1][1]:
            # interval[0]  is the starting point of interval
            # result[-1][1] is "3" of interval [1,3]

            # Merge them - as overlap
            result[-1][1] = max(result[-1][1], interval[1])

        else:

            # No overlap
            result.append(interval)

    return result

intervals = [[1,3], [2,6], [8,10], [15,18]]
print(merge_intervals(intervals))
```
