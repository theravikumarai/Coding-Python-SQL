```python
def frequency_sort(nums):
    frequency = {}

    # Count frequency of each number
    for num in nums:
        frequency[num] = frequency.get(num, 0) + 1

    # Sort numbers according to their frequency
    return sorted(nums, key=lambda x: frequency[x], reverse=True)


nums = [4, 6, 2, 6, 4, 4, 1, 2]

print(frequency_sort(nums))


# Expected: [4, 4, 4, 6, 6, 2, 2, 1]
```
