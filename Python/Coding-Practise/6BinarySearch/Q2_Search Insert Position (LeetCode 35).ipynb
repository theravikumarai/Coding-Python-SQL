This is one of the most important Binary Search questions because it teaches the concept of finding a position even when the element doesn't exist.

```python
def search_insert(nums, target):
    left = 0
    right = len(nums) - 1

    while left <= right:
        mid = (left + right) // 2

        if nums[mid] == target:
            return mid

        elif nums[mid] < target:
            left = mid + 1

        else:
            right = mid - 1
    
    return left

nums = [-3, -1, 2, 5, 9]
target = 3

print(search_insert(nums, target))
```
