```python
def last_occurrence(nums, target):

    left = 0
    right = len(nums) - 1

    answer = -1

    while left <= right:

        mid = (left + right) // 2

        if nums[mid] == target:
            answer = mid
            left = mid + 1

        elif nums[mid] < target:
            left = mid + 1

        else:
            right = mid - 1

    return answer

nums = [-3, -1, 2, 2,2, 5, 9]
target = 2
print(last_occurrence(nums, target))
```
