```python
def first_occurrence(nums, target):

    left = 0
    right = len(nums) - 1
    answer = -1

    while left <= right:

        mid = (left + right) // 2

        if nums[mid] == target:
            answer = mid
            right = mid - 1

        elif nums[mid] < target:
            left = mid + 1

        else:
            right = mid - 1

    return answer


def last_occurrence(nums, target):

    left = 0
    right = len(nums) - 1
    answer = -1

    while left <= right:

        mid = (left + right) // 2

        if nums[mid] == target:
            answer = mid
            left = mid + 1

        elif nums[mid] < target:
            left = mid + 1

        else:
            right = mid - 1

    return answer


def count_occurrences(nums, target):

    first = first_occurrence(nums, target)

    if first == -1:
        return 0

    last = last_occurrence(nums, target)
    count = last - first + 1

    return count


nums = [-3, 1, 2,4, 4, 4, 4, 6]
target = 4
print(count_occurrences(nums, target))
```
