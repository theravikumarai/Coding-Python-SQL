LeetCode: 69 - Sqrt(x)



```python
def sqrt(x=5):
    i = 1

    while i*i <= x:
        i += 1

    return i-1

print(sqrt())
```

```python
def my_sqrt(x):
    left = 0
    right = x
    result = 0

    while left <= right:

        mid = (left + right) // 2

        if mid*mid == x:
            return mid

        elif mid * mid < x:
            result = mid
            left = mid + 1

        elif mid * mid > x:
            right = mid - 1

    return result






# Test Cases
print(my_sqrt(16))   # 4
# print(my_sqrt(20))   # 4
# print(my_sqrt(36))   # 6
# print(my_sqrt(1))    # 1
# print(my_sqrt(0))    # 0
```
