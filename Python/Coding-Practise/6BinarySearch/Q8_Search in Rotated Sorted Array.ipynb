### How to Check if Left Half is Sorted?

    Suppose
    4 5 6 7 0 1 2
    L      M      R

    If
    nums[left] <= nums[mid]

    then
    Left half is sorted.

    Example
    4 <= 7

    True.

Otherwise

    If

    nums[left] > nums[mid]

    then

    Right half is sorted.

```python
def search_rotated(nums, target):
    left = 0
    right = len(nums)- 1
    result = 0

    while left <= right:
        mid = (left + right) // 2

        if nums[mid] == target:
            return mid

        elif nums[left] <= nums[mid]:

            if nums[left] <= target < nums[mid]:
                right = mid - 1

            else:
                left = mid + 1


        else:

            if nums[mid] < target <= nums[right]:
                left = mid + 1

            else:
                right = mid - 1

    return -1


nums = [4,5,6,7,0,1,2]
target = 0
print(search_rotated(nums, target))
```
