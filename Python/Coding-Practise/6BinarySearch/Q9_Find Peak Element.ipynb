    Notice the differences from previous Binary Search problems:

    We use while left < right, not <=.
    We compare nums[mid] with nums[mid + 1].
    We return left (or right, since both are equal at the end).

```python
def find_peak_element(nums):

    left = 0
    right = len(nums)-1

    while left < right:

        mid = (left + right)//2

        if nums[mid] > nums[mid+1]:

            right = mid

        else:

            left = mid + 1

    return left

nums = [1,3,4,5,6,7,1]
print(find_peak_element(nums))

# Return Index value
```
