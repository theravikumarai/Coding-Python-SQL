# Binary Search Template (Memorize)

    left = 0
    right = len(nums) - 1

    while left <= right:
        mid = (left + right) // 2

        if nums[mid] == target:
            return mid

        elif nums[mid] < target:
            left = mid + 1

        else:
            right = mid - 1

    return -1

    Every Binary Search problem starts from this template.

## Time Complexity

    Suppose, n = 16

    After each iteration:

    16
    ↓
    8
    ↓
    4
    ↓
    2
    ↓
    1

    Number of iterations:
    log₂(16) = 4

    General formula:
    O(log n)

## Space Complexity

    We only use:

    left
    right
    mid

    No extra array or dictionary.

O(1)

# Common Mistakes

    Mistake 1

    Using Binary Search on an unsorted array.

    Mistake 2

    Using
    while left < right

    instead of
    while left <= right

    Mistake 3

    Writing
    left = mid

    instead of
    left = mid + 1

    This can cause an infinite loop.

    Mistake 4

    Writing
    right = mid

    instead of
    right = mid - 1

    Again, this can cause an infinite loop.
