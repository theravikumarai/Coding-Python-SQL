````python
# Two Pointers:

def reverse_string(s):
    chars = list(s)   # string are immutable

    left = 0          #  point begining of first character
    right = len(chars) - 1    #point last character

    while left < right:
        chars[left], chars[right] = chars[right], chars[left]
        left +=1  
        right -=1

    return "".join(chars)
````
