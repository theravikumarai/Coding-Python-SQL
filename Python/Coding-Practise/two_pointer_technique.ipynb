# Types of two pointers:

# General Algorithm

    Initialize left pointer
    initialize right pointer

    while condition:
        Evaluate current state

        if answer found:
            return

        Move one or both pointers according to the problems.

The key idea is that you never move a pointer backward unnecessarily, so each pointer usually visits each element at most once.

# How to Identift Two Pointer Problems

    Look for clues like:

    Sorted array
    Pair of elements
    Triplets
    Remove duplicates
    Reverse array/string
    Palindrome
    Merge arrays
    Consecutive elements
    Subarray/substring

## Advantages

    Reduces time complexity from O(n²) to O(n) in many problems.
    Uses O(1) extra space in most cases.
    Produces clean, efficient solutions.
