```python
# # Find email from text   -----> First Match

text = "Contact us at support@gmail.com or raviikrds@gmail.com"

match = re.search(r"\S+@\S+", text)

# print(match)

if match:
    print(match.group())

# # It searches the entire text until it finds the first match.

# # Intuition
# # re.search() scans entire text until it finds first match.

# Regex Symbols Explained:
# \S Means: non-space character
# + Means: one or more
# So:
# \S+@\S+ means: something@something
```

```python
# MOST IMPORTANT REGEX SYMBOLS

# Pattern	Meaning
# .	    Any character
# \d	Digit
# \D	Non-digit
# \w	Word character
# \W	Non-word
# \s	Space
# \S	Non-space
# +	    One or more
# *	    Zero or more
# ?	    Optional
# {n}	Exactly n
# {m,n}	Between m and n
# ^	    Start
# $	    End
```

## Real Practical Senarios

```python
# Extract error log

import re
logs = """
INFO Server started
ERROR Database failed
INFO User login
ERROR Timeout occurred
Error: No network 
"""

errors = re.findall(r"Error.*", logs, flags= re.IGNORECASE)   # dot means any character and * means all conternt of lines. flags= re.IGNORECASE ---> upper case & lower case letters are  treated as identical pairs
print(errors)
```

```python
# Extract URLs

import re
text = """
Visit https://google.com
Also check https://openai.com
also check https://www.youtube.com/results?search_query=ai+research+scientist
"""
urls = re.findall(r"https?://\S+", text)   #here s is optional, it means it can check for http as well.  xyz(ab)? here both ab optional
print(urls)

```

```python
# Validate Emails

import re
email = "test@gmail.com"
pattern = r"^[\w\.-]+@[\w\.-]+\.\w+$"

if re.fullmatch(pattern, email):
    print("Valid")

else:
    print("Invalid")


# ^ Start: Matches the very beginning of 
# the text.[\w\.-]+ Username: Matches one or 
# more letters, numbers, underscores, dots, 
# or hyphens.@ The At Sign: Matches the exact @ symbol.[\w\.-]+ Domain Name: Matches one or more letters, 
# numbers, dots, or hyphens for the mail server.\. The Dot: Matches the exact literal dot . before the extension.\w+$ Extension & End: Matches the ending letters (like com or org) right up to the very end of the text.
```

```python
# Extract Dates

import re

text = "Meeting on 28-05-2026"
a = "Meeting on 2026/05/02"

date = re.findall(r"\d{2}-\d{2}-\d{4}", text)
d = re.findall(r"\d{4}/\d{2}/\d{2}", a)
print(date,d)
```

```python
# Chat Moderation System
import re

message = "You are idiot and stupid"

bad_words = ["idiot", "stupid", "crazy"]

# \b makes sure it matches the whole word, not just part of a word
pattern = r"\b(" + "|".join(bad_words) + r")\b"

# flags=re.IGNORECASE catches IDIOT, Idiot, and idiot
cleaned = re.sub(pattern, "***", message, flags=re.IGNORECASE)
print(pattern)
print(type(pattern))
print(cleaned)
```

```python
# WIth compile ---> by creating regex object
import re
emails = ['raviikrds@gmail.com', 'iamravi@gmail.com', 'a@outlook.com']
email_pattern = re.compile(r"\S+@\S+")   # creates regex object

for email in emails:
    match = email_pattern.search(email)
    if match:
        print(f"Matched: {match.group()}")


# Now Python:

# Parses the pattern once
# Creates a reusable regex object
# Reuses it thousands/millions of times

# This is faster and cleaner.
```

```python
# With compile
import re
text1 = "Age is 25"
text2 = "My salary is 10000000"

number_pattern = re.compile(r"\d+")   #regex object

print(number_pattern.search(text1).group())
print(number_pattern.search(text2).group())
```

```python
# Email Validation

import re

email_pattern = re.compile(
    r"^[\w\.-]+@[\w]+\.\w+$"
)
# [\w\.-]+: This requires at least one letter, number, dot, or dash for the username and domain.    \.\w+: This forces a real dot followed by letters for the extension (like .com or .in).
emails = [
    "abc@gmail.com",
    "wrong-email",
    "john@yahoo.com"
]

for email in emails:
    if email_pattern.fullmatch(email):
        print("Valid:", email)
    # else:
    #     print("Invalid")
```

## Real Industry senario:

```python
# Suppose millions of logs arrives
logs = [
    "INFO User Login",
    "ERROR DB Failed",
    "INFO Connected",
    "ERROR Timeout"
]
# Create pattern once
import re

error_pattern = re.compile(r"ERROR.*") #"Prepare this regex rule once, then use it wherever needed."

# use repeatidily

for log in logs:
    if error_pattern.search(log):
        print(log)
    else:
        pass
```

# Common FLags
| Flag            | Meaning                   |
| --------------- | ------------------------- |
| `re.IGNORECASE` | Ignore upper/lower case   |
| `re.MULTILINE`  | `^` and `$` work per line |
| `re.DOTALL`     | `.` matches newline       |
| `re.VERBOSE`    | Write readable regex      |
