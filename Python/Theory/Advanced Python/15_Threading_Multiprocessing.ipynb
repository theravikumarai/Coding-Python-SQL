🔹 What is Threading?

Threading in Python allows multiple threads to execute concurrently within the same process, mainly improving performance for I/O-bound tasks.

🔹 Why use Threading?

To improve responsiveness

To run tasks in parallel (especially I/O-bound tasks)

To avoid blocking the main program

🔹 Key Concepts

Thread: A lightweight unit of execution

Main Thread: Default thread where program starts

GIL (Global Interpreter Lock): Allows only one thread to execute Python bytecode at a time
→ Threads are best for I/O-bound, not CPU-bound tasks

````python
import threading
import time

def walk_dog():
    time.sleep(8)
    print("You finish walking the dog")

def take_out_trash():
    time.sleep(2)
    print("You get the mail")

def get_mail():
    time.sleep(4)
    print("You get the Mail")

walk_dog()
take_out_trash()
get_mail()

## Above function running on the same thread, so it will execute one by one upon calling.
````

````python
import threading
import time

def walk_dog(first, last):
    time.sleep(8)
    print(f"You finish walking {first} {last}")

def take_out_trash():
    time.sleep(2)
    print("You throw the trash")

def get_mail():
    time.sleep(4)
    print("You get the Mail")

# Constructing a thread object
a = threading.Thread(target=walk_dog, args=("Tony", "Tommy"))
a.start()

b = threading.Thread(target=take_out_trash)
b.start()

c = threading.Thread(target=get_mail)
c.start()

## Now the above function execute concurrently(Same Time). Unorder

## We can use join() method for executing in proper order
a.join()
b.join()
c.join()

print('All tasks are completed!')
````

````python
# 🔹 Important Methods
# Method	Description

# start()	Starts the thread
# run()	Code executed by thread
# join()	Waits for thread to finish
# is_alive()	Checks if thread is running
````

# Multiprocessing:

🔹 What is Multiprocessing?

Multiprocessing allows a Python program to run multiple processes in parallel, each with its own memory space and Python interpreter.

👉 This bypasses the GIL (Global Interpreter Lock), making multiprocessing ideal for CPU-bound tasks.

🔹 Why use Multiprocessing?

To utilize multiple CPU cores

To speed up heavy computations

To avoid GIL limitations



````python
# Process vs Thread (Quick View)
| Feature    | Threading  | Multiprocessing |
| ---------- | ---------- | --------------- |
| Execution  | Concurrent | Parallel        |
| Memory     | Shared     | Separate        |
| GIL effect | Yes        | ❌ No            |
| Best for   | I/O-bound  | CPU-bound       |
| Overhead   | Low        | Higher          |
````

````python
🔹 When to Use Multiprocessing?

✅ Best for:

Data processing
Machine Learning preprocessing
Mathematical computations
Image processing

❌ Avoid for:
Simple I/O tasks (use threading instead)

🔹 Interview One-Line Definition
Multiprocessing in Python allows true parallel execution by running tasks in separate processes, making it ideal for CPU-bound operations.
````

````python
#Example

from multiprocessing import Process
import time

def square(nums):
    for n in nums:
        time.sleep(1)
        print(n * n)

nums = [1, 2, 3, 4]

p1 = Process(target=square, args=(nums,))
p2 = Process(target=square, args=(nums,))

p1.start()
p2.start()

p1.join()
p2.join()

print("Done")
````

````python
from multiprocessing import Pool
import math

def compute(n):
    return math.factorial(n)

if __name__ == "__main__":
    with Pool() as p:
        print(p.map(compute, [100000, 200000, 300000]))
````

````python
## 🚀 Real-World ML Use Case of Multiprocessing (Using CPU Cores)

Let’s take a **practical ML scenario** you might face in industry 👇

---

# 📌 Use Case: Parallel Data Preprocessing for ML

### 🎯 Problem

You have:

* 5 lakh images
* Need to resize, normalize, and extract features
* Each image processing takes CPU time

If you process sequentially → Very slow ❌
If you use multiprocessing → Much faster ✅

---

## 🔹 Why Multiprocessing in ML?

In ML pipelines:

* Feature engineering
* Image processing
* Text preprocessing
* Data cleaning
* Heavy mathematical transformations

These are **CPU-bound tasks** → Perfect for multiprocessing.

---

# 🧠 Step 1: Check Number of CPU Cores

```python
import multiprocessing

print("Number of CPU cores:", multiprocessing.cpu_count())
```

Example Output:

```
Number of CPU cores: 8
```

---

# 🧪 Example: Parallel Feature Processing

### 🎯 Scenario:

We simulate heavy computation (like feature extraction)

```python
from multiprocessing import Pool, cpu_count
import time

def feature_engineering(x):
    time.sleep(1)   # Simulating heavy computation
    return x * x

if __name__ == "__main__":
    
    data = [1,2,3,4,5,6,7,8]

    cores = cpu_count()
    print("Using cores:", cores)

    with Pool(cores) as p:
        results = p.map(feature_engineering, data)

    print(results)
```

---

## 🔥 What Happens?

* If you have 8 cores → 8 tasks run in parallel
* Instead of 8 seconds → ~1 second
* Massive performance improvement

---

# 📊 Real ML Example: Text Preprocessing

Imagine:

* Cleaning 1 million reviews
* Removing stopwords
* Lemmatization
* Tokenization

Instead of:

```python
for review in reviews:
    process(review)
```

Use:

```python
with Pool(cpu_count()) as p:
    cleaned_reviews = p.map(process, reviews)
```

✔ Fully utilizes CPU
✔ Speeds up preprocessing stage

---

# 🖼️ Real ML Example: Image Dataset Preprocessing

![Image](https://educative.io/api/edpresso/shot/5919342641020928/image/5073097894658048.png)

![Image](https://cdn.botpenguin.com/assets/website/Feature_Extraction_14fa61bcea.webp)

![Image](https://www.teach-ict.com/as_as_computing/ocr/H447/F453/3_3_3/parallel_processors/miniweb/images/multi_cpu.gif)

![Image](https://researchcomputing.princeton.edu/sites/g/files/toruqf7796/files/styles/freeform_750w/public/2021-11/parallelProblem_hpc.llnl_.gov_.gif?itok=xdVlU5oO)

In real-world:

1. Load image
2. Resize
3. Normalize
4. Extract features
5. Save processed output

Each image → Independent task
Perfect for multiprocessing.

---

# 📌 Advanced: Limit Number of Cores

Sometimes you don't want to use all cores.

```python
with Pool(4) as p:   # Use only 4 cores
    results = p.map(feature_engineering, data)
```

Why?

* System stability
* Running other services
* Avoid 100% CPU usage

---

# 🏢 Real Industry Example

In companies:

* Fraud detection models preprocess transactions
* NLP models clean millions of texts
* Recommendation systems compute embeddings
* Computer Vision pipelines process images

All these use:

* `multiprocessing`
* `joblib`
* `dask`
* Spark (for distributed computing)

---

# 🎯 Interview Answer (Short Version)

> In real-world ML, multiprocessing is commonly used during data preprocessing and feature engineering stages. Since these are CPU-intensive tasks, using multiprocessing with `Pool(cpu_count())` allows us to fully utilize all available cores and significantly reduce training pipeline time.

---

# 🧠 When NOT to Use Multiprocessing in ML?

* When using GPU training (TensorFlow / PyTorch handle parallelism internally)
* Small datasets
* I/O-heavy tasks (use threading instead)

---

If you want next:

* 🔥 Joblib parallel processing in ML
* 🔥 Dask vs Multiprocessing
* 🔥 How scikit-learn uses `n_jobs`
* 🔥 End-to-end ML pipeline optimization strategy
````
