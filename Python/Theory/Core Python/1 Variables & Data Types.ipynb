# Python Variables & Data Types

## 1. Variable

A **variable** is a name that refers to a value stored in memory.

```python
name = "Ravi"
age = 25
salary = 50000.0
```

**Key Points**

* No type declaration required (Dynamic Typing)
* Variables store **references** to objects
* Use meaningful names

---

# 2. Variable Naming Rules

✅ Valid

```python
user_name = "Ravi"
age2 = 25
_learning_rate = 0.001
```

❌ Invalid

```python
2age = 25
class = "AI"
first-name = "Ravi"
```

**Best Practice (PEP 8)**

```python
learning_rate
batch_size
train_data
model_accuracy
```

Use **snake_case** for variable names.

---

# 3. Dynamic Typing

Python automatically detects the data type.

```python
x = 10
x = "Hello"
x = 3.14
```

No explicit type declaration is needed.

# 16. Mutable vs Immutable ⭐⭐⭐⭐⭐

## Mutable

Objects can be modified.

* list
* dict
* set

Example

```python
nums.append(5)
```

---

## Immutable

Objects cannot be modified.

* int
* float
* str
* tuple
* bool
* None

Example

```python
name = "Ravi"

# Not Allowed
name[0] = "A"
```

---

# 17. Reference Concept ⭐⭐⭐⭐

```python
a = 10
b = a

a = 20

print(b)
```

Output

```python
10
```

Variables hold references to objects, not copies by default.
