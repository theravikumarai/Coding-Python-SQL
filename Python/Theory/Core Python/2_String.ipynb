```python
## String Functions
# 1. len() - Returns the length of the string
# 2. str() - Converts a value to a string
# 3. upper() - Converts all characters in the string to uppercase
# 4. lower() - Converts all characters in the string to lowercase
# 5. strip() - Removes leading and trailing whitespace from the string
# 6. replace(old, new) - Replaces occurrences of 'old' substring with 'new' substring
# 7. split(separator) - Splits the string into a list of substrings based on the specified separator
# 8. join(iterable) - Joins elements of an iterable (like a list    ) into a single string, separated by the string on which it is called eg: ','.join(['a', 'b', 'c']) -> 'a,b,c'
# 9. find(substring) - Returns the lowest index of the substring if found in the string, otherwise returns -1
# 10. format() - Formats the string using placeholders   eg: text = "Hello, {}"  ----> text.format("World") ----> "Hello, World"
# 11. endswith(suffix) - Returns True if the string ends with the specified suffix, otherwise False eg: "hello.py".endswith(".py") -> True
# 12. startswith(prefix) - Returns True if the string starts with the specified prefix, otherwise False eg : "hello.py".startswith("he") -> True
# 13. capitalize() - Capitalizes the first character of the string eg : "hello ravi".capitalize() -> "Hello ravi"
# 14. isdigit() - Returns True if all characters in the string are digits, otherwise False  eg: "12345".isdigit() -> True
# 15. isalpha() - Returns True if all characters in the string are alphabetic eg : "Hello".isalpha() -> True
# 16. count(substring) - Returns the number of occurrences of the substring in the string eg: "hello world".count("o") -> 2
# 17. index(substring) - Returns the lowest index of the substring if found in the string, raises ValueError if not found
# 18. reverse() - Reverses the string (can be done using slicing: string[::-1]) eg: "hello"[::-1] -> "olleh"
# 19. title() - covert string into title
# 20. copy() - Creates a copy of the string (strings are immutable, so this is often unnecessary) eg: original = "hello"; copy_str = original[:]
```

```python
# Your raw text containing the INC number
import re
text = "The issue was logged under ticket REQ590454534534543 and sent to support."
# Search for 'inc' followed by digits
# [0-9]+ means match one or more numbers
match = re.search(r'REQ[0-9]+', text, re.IGNORECASE)
print(match.group())
```

```python
text = "The issue was logged under ticket REQ590454fdg%&ds534534543 and sent to support."
match = re.search(r'REQ\S+', text, re.IGNORECASE)
print(match.group())
print(re.findall(r'\d+',text))
```
