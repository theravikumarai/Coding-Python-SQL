# Python Lists:

---

# 1. What is a List?

* A **list** is an ordered, mutable collection of items.
* It can store multiple values in a single variable.
* Lists are enclosed using `[]`.

```python
numbers = [10, 20, 30, 40]
```

---

# 2. Why Lists are Important in AI/ML

Lists are the first step in data processing.

```text
Raw Data
    ↓
Python List
    ↓
NumPy Array
    ↓
Pandas DataFrame
    ↓
Tensor (PyTorch/TensorFlow)
    ↓
Machine Learning Model
```

Used for:

* Storing datasets
* Labels
* Features
* Predictions
* Sensor readings
* Text tokens (before preprocessing)

# 4. Indexing

Python uses **0-based indexing**.

```python
nums = [10,20,30,40]

nums[0]      # 10
nums[2]      # 30
```

---

# 5. Negative Indexing

Access elements from the end.

```python
nums = [10,20,30,40]

nums[-1]     # 40
nums[-2]     # 30
```

---

# 6. Slicing

Extract part of a list.

```python
nums = [10,20,30,40,50]

nums[1:4]
```

Output

```python
[20,30,40]
```

Useful for:

* Train/Test split
* Batch creation
* Feature selection

---

# 7. List is Mutable

Elements can be changed.

```python

```

Result


# 12. enumerate()

Used when both **index** and **value** are needed.

```python
names = ["A","B","C"]

for index, value in enumerate(names):
    print(index, value)
```

Output

```python
0 A
1 B
2 C
```

Start from 1

```python
for i, value in enumerate(names, start=1):
    print(i, value)
```

Output

```python
1 A
2 B
3 C
```

AI Use:

* Row numbers
* Prediction IDs
* Dataset debugging

# 13. List Comprehension ⭐

Pythonic way of creating lists.

Instead of

```python
squares = []

for x in range(5):
    squares.append(x*x)
```

Use

```python
squares = [x*x for x in range(5)]
```

Output

```python
[0,1,4,9,16]
```

AI Use:

* Data preprocessing
* Feature engineering
* Normalization
* Text cleaning

# 18. Common Interview Questions

### Difference between append() and extend()

```python
a = [1,2]

a.append([3,4])

# [1,2,[3,4]]
```

```python
a = [1,2]

a.extend([3,4])

# [1,2,3,4]
```

---

### Difference between remove() and pop()

`remove()`

* Removes by value

```python
nums.remove(20)
```

`pop()`

* Removes by index

```python
nums.pop(1)
```

---

### Difference between sort() and sorted()

`sort()`

* Changes original list

```python
nums.sort()
```

`sorted()`

* Returns a new sorted list

```python
new = sorted(nums)
```

````python
## Enumerate : enumerate() is a very useful function when you need both the index and the value while looping through a list.
# Without Enumerate
names = ["Ravi", "Rahul", "Rishu", "Rishav"]

for i in range(len(names)):
    print(i+1, names[i])

````

````python
# Using Enumerate

names = ["Ravi", "Rahul", "Rishu", "Rishav"]
for index, name in enumerate(names, start=1):
    print(index, name)
````

````python
fruits = ["apple", "banana", "cherry"]

for idx, fruit in enumerate(fruits, start =1):
    print(idx, fruit)
````

````python
### for example, some predictions are wrong
prediction = [1,0,1,0]
actual = [1,1,1,0]
for i, pred in enumerate(prediction):
    if pred != actual[i]:
        print(f"wrong prediction at row {i+1}")
````

````python
# Using zip()  ---> zip pairs up each prediction and actual value together

prediction = [1,0,3,0]
actual = [1,1,1,0]
# zip pairs up each prediction and actual value together
for i, (pred,act) in enumerate(zip(prediction, actual)):
    if pred != act:
        print(f" Worng prediction at row {i+1}")
````

````python
a = [1,1,1,1,1]
b = [1,2,1,1,1]
for k, (i,j) in enumerate(zip(a,b)):
    if i != j:
        print(f"no match at {k}")
````

````python
a ="ravi"
b= "rava"
for i, (j,k) in enumerate(zip(a,b)):
    if j != k:
        print(f"word not matched at {i}")
````

````python
output1 = [1,0,1,1,1,0]
output2 = [1,1,0,0,0,1]

for idx,(res1,res2) in enumerate(zip(output1,output2), start=1):
    if res1 != res2:
        print(f"output not equal at {idx}")
````
