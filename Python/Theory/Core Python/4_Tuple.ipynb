# 1. What is a Tuple?

A **tuple** is an **ordered, immutable collection** of items.

* Ordered → Maintains insertion order.
* Immutable → Cannot be modified after creation.
* Allows duplicate values.
* Can store multiple data types.

```python
t = (1, 2, 3)
```

**Memory Tip**

> **Tuple = Fixed Data**
> Once created, it cannot be changed.

# 20. Important Interview Questions

### Difference between List and Tuple?

* List is mutable.
* Tuple is immutable.
* Tuple is faster and uses less memory.
* Lists are used for changing data.
* Tuples are used for fixed data.

---

### Why are tuples immutable?

To protect important information from accidental modification and improve performance.

---

### Why can tuples be dictionary keys?

Because they are immutable and therefore hashable (provided all their elements are also immutable).

---

### Can a tuple contain mutable objects?

Yes.

```python
t = ([1,2],3)
```

The tuple itself cannot change, but the list inside it can.

```python
t[0].append(4)
```

Output

```
([1,2,4],3)
```

---

### Does `return a, b` return two values?

No.

It returns **one tuple** containing two values.

# Quick Revision

* **Tuple** = Ordered + Immutable collection.
* Created using `()`.
* Single-element tuple requires a trailing comma: `(5,)`.
* Supports indexing, slicing, iteration, concatenation, and repetition.
* Only **2 methods**: `count()` and `index()`.
* Can return multiple values from functions (actually returned as a tuple).
* Supports packing and unpacking.
* Can be used as dictionary keys (if all elements are immutable).
* Common AI/ML uses: tensor shapes, image sizes, coordinates, RGB values, dataset samples, hyperparameters, and function return values.
* **Choose a tuple when the data should never change; choose a list when the data needs to be modified.**

Tuple Methods

Tuples have only two methods because they are immutable.

1. count()                     2. index()

````python
#Tuple Methods
a = (1,2,3,1,4,5,1)
print(a.count(1))  # Counts how many times 1 appears in the tuple
print(a.index(4))  # Finds the index of the first occurrence of 4
print(a.index(1,2))  # Finds the index of the first occurrence of 1 starting from index 2
print(a.index(4,1)) 
````

Most Important AI/ML Intuition

When you see a tuple in AI/ML code, it usually means:

"This is a fixed piece of information that should not change"

Examples:

image_shape = (224, 224, 3)

train_test_split = (8000, 2000)

coordinate = (x, y)

(loss, accuracy)

These are all records with a fixed structure, which is exactly where tuples are most useful.
