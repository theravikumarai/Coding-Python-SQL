# Python Dictionary

## Definition

A **Dictionary** is a mutable collection that stores data as **Key → Value** pairs.

```python
person = {
    "name": "Ravi",
    "age": 25
}
```

```
"name" → "Ravi"
"age"  → 25
```

# Word Frequency (Most Important AI Example)

```python
text = "cat dog cat cat"

freq = {}

for word in text.split():
    freq[word] = freq.get(word, 0) + 1

print(freq)
```

Output

```python
{
    "cat":3,
    "dog":1
}
```

Why `get()`?

```
cat exists?
     ↓
No
Return 0

0 + 1 = 1
```

Next time

```
cat exists?
     ↓
Yes

Return 1

1 + 1 = 2
```

## Dictionary → JSON File

### `json.dump()`


---

## JSON File → Dictionary

### `json.load()`

```python

```

---

# Remember These Four Functions

| Function  | Input       | Output      |
| --------- | ----------- | ----------- |
| `loads()` | JSON String | Dictionary  |
| `dumps()` | Dictionary  | JSON String |
| `load()`  | JSON File   | Dictionary  |
| `dump()`  | Dictionary  | JSON File   |

### Memory Trick

```
s = String

loads()  → String → Dictionary
dumps()  → Dictionary → String

No s = File

load()   → File → Dictionary
dump()   → Dictionary → File
```

# Interview Questions

### Q1. Why use `get()` instead of `[]`?

* `[]` raises `KeyError` if the key is missing.
* `get()` safely returns `None` or a default value.

---

### Q2. Difference between Dictionary and JSON?

| Dictionary         | JSON                               |
| ------------------ | ---------------------------------- |
| Python object      | Text format                        |
| Type = `dict`      | Type = `str`                       |
| Used inside Python | Used for storage and communication |

---

### Q3. Difference between `load()` and `loads()`?

* `load()` → Reads **JSON file**.
* `loads()` → Parses **JSON string**.

---

### Q4. Difference between `dump()` and `dumps()`?

* `dump()` → Writes **JSON file**.
* `dumps()` → Returns **JSON string**.

````python
## Dictionary Methods:
.keys() - Returns a view object that displays a list of all the keys in the dictionary.
.values() - Returns a view object that displays a list of all the values in the dictionary.
.items() - Returns a view object that displays a list of dictionary's key-value tuple pairs.
.get() - Returns the value for a specified key if the key is in the dictionary.
.update() - Updates the dictionary with elements from another dictionary object or from an iterable of key/value pairs.
sorted() - Returns a sorted list of the dictionary's keys.
setdefault() - Returns the value of a key if it is in the dictionary; if not, it inserts the key with a specified value.
````

````python
# .keys(): Returns a dict_keys object, which is a view of all keys in the dictionary. This can be converted to a list.

student = {"name": "Rahul Kumar", "subjects": {"Physics": 90}}
all_keys = student.keys()
print(all_keys)
# Output: dict_keys(['name', 'subjects'])

# Convert to a list
print(list(all_keys))
# Output: ['name', 'subjects']

# Get the number of key-value pairs
print(len(student))
# Output: 2
print(len(list(student.keys())))
# Output: 2
````

````python
# .values(): Returns a dict_values object, a view of all values in the dictionary. This can be converted to a list.

student = {"name": "Rahul Kumar", "subjects": {"Physics": 90}}
all_values = student.values()
print(all_values)
# Output: dict_values(['Rahul Kumar', {'Physics': 90}])
print(type(all_values))
# Convert to a list
print(list(all_values))
# Output: ['Rahul Kumar', {'Physics': 90}]
````

````python
# .items(): Returns a dict_items object, a view of all key-value pairs as tuples. This can be converted to a list of tuples.

student = {"name": "Rahul Kumar", "subjects": {"Physics": 90}}
all_items = student.items()
print(all_items)
# Output: dict_items([('name', 'Rahul Kumar'), ('subjects', {'Physics': 90})])

# Convert to a list of tuples
list_of_pairs = list(all_items)
print(list_of_pairs)
# Output: [('name', 'Rahul Kumar'), ('subjects', {'Physics': 90})]

# Access individual pairs
print(list_of_pairs[0]) # Output: ('name', 'Rahul Kumar')
print(list_of_pairs[1]) # Output: ('subjects', {'Physics': 90})
````

````python
# .get(key, default_value=None): Returns the value for the specified key. If the key is not found, it returns None (or a specified default_value) instead of raising a KeyError. This is crucial for robust error handling in real-world applications, as it prevents program crashes when keys might be missing.

student = {"name": "Rahul Kumar"}
print(student.get("name"))     # Output: Rahul Kumar
print(student.get("surname"))  # Output: None (no error)

# Comparison with direct access
# print(student["surname"]) # This would cause a KeyError and stop execution

# Demonstrating error handling
print("Before error")
print(student.get("non_existent_key")) # Does not raise an error
print("After get() method")

# print("Before error")
# print(student["non_existent_key"]) # This line would cause an error
# print("After direct access") # This line would not be executed due to the error

# When a KeyError occurs, any code following it in the same block will not execute. Using .get() helps prevent this by gracefully handling missing keys.
````

````python
# .update(other_dictionary): Adds elements from another dictionary (or an iterable of key-value pairs) into the current dictionary. If a key from other_dictionary already exists in the current dictionary, its value is updated. New keys are added.

student = {"name": "Rahul Kumar", "age": 20}
student.update({"city": "Delhi"})
print(student)
# Output: {'name': 'Rahul Kumar', 'age': 20, 'city': 'Delhi'}

student.update({"age": 21, "gender": "Male"})
print(student)
# Output: {'name': 'Rahul Kumar', 'age': 21, 'city': 'Delhi', 'gender': 'Male'}
# 'age' value is updated from 20 to 21, 'gender' is added.
````
