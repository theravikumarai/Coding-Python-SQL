# 1. What are Conditional Statements?

Conditional statements allow a program to **make decisions** by executing different blocks of code based on whether a condition is `True` or `False`.

**Real-life analogy:**

> **If** it is raining → Carry an umbrella.
>
> **Else** → Wear sunglasses.

**In AI/ML:**

* Validate data
* Clean datasets
* Select models
* Convert probabilities to predictions
* Stop training
* Feature engineering
* Business rule implementation

# 10. Ternary Operator (Conditional Expression)

A compact way to write a simple `if-else`.

### Syntax

```python
value_if_true if condition else value_if_false
```

Example

```python
status = "Adult" if age >= 18 else "Minor"
```

Equivalent

```python
if age >= 18:
    status = "Adult"
else:
    status = "Minor"
```

---


# 11. Match-Case Statement (Python 3.10+)

Used to compare **exact values** (similar to switch-case in other languages).

### Syntax

```python
match variable:

    case value1:
        ...

    case value2:
        ...

    case _:
        ...
```

`_` acts as the default case.

Example

```python
match algorithm:

    case "RandomForest":
        print("Tree Model")

    case "XGBoost":
        print("Boosting")

    case _:
        print("Unknown")
```

# 15. Which Statement Should You Use?

| Situation                    | Best Choice      |
| ---------------------------- | ---------------- |
| One condition                | `if`             |
| Two choices                  | `if-else`        |
| Multiple ranges/conditions   | `if-elif-else`   |
| One condition inside another | Nested `if`      |
| Simple assignment            | Ternary operator |
| Many exact values            | `match-case`     |

---

# 16. Best Practices

✅ Keep conditions simple.

```python
if accuracy > 90:
```

---

✅ Use `elif` instead of many separate `if` statements when only one branch should execute.

---

✅ Use meaningful variable names.

```python
model_accuracy
```

Instead of

```python
a
```

---

✅ Use parentheses for long conditions.

```python
if (
    accuracy > 90
    and f1_score > 85
):
```

---

✅ Use a ternary operator only for **simple assignments**.

Good

```python
status = "Pass" if score >= 40 else "Fail"
```

Avoid

```python
result = (
    "A" if score > 90
    else "B" if score > 80
    else "C" if score > 70
    else "D" if score > 60
    else "F"
)
```

Use `if-elif-else` instead.

# 18. Interview Questions

### Difference between `if` and `if-else`?

* `if` executes only when the condition is `True`.
* `if-else` always executes one of two branches.

---

### When should you use `elif`?

When there are multiple **mutually exclusive** conditions.

---

### When should you use `match-case`?

When comparing a variable against many **exact values**.

---

### What is a ternary operator?

A one-line conditional expression for simple assignments.

```python
value_if_true if condition else value_if_false
```

---

### Can ternary operators replace all `if-else` statements?

No. They are intended for **simple expressions**. Use standard `if-elif-else` for complex logic.

---

### What is the difference between `if-elif-else` and `match-case`?

| `if-elif-else`                                  | `match-case`                                            |
| ----------------------------------------------- | ------------------------------------------------------- |
| Evaluates conditions and ranges                 | Matches exact values or patterns                        |
| Supports logical operators (`and`, `or`, `not`) | Primarily matches literal values or structural patterns |
| Most common in AI/ML                            | Less common in AI/ML                                    |

````python
## Calculator
def calculator():
    print("--- Python Match-Case Calculator ---")
    
    while True:
        operator = input("\nSelect an operator (+, -, *, /) or 'q' to exit: ").strip()
        
        if operator.lower() == 'q':
            break
        if operator not in ['+', '-', '*', '/']:
            print("Invalid operator!")
            continue
            
        try:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
        except ValueError:
            print("Invalid number input!")
            continue
            
        match operator:
            case "+": print(f"Result of {num1} and {num2}: {num1 + num2}")
            case "-": print(f"Result of {num1} and {num2}: {num1 - num2}")
            case "*": print(f"Result of {num1} and {num2}: {num1 * num2}")
            case "/":
                if num2 != 0: print(f"Result of {num1} and {num2}: {num1 / num2}")
                else: print("Error: Division by zero!")
            case _: print("Unexpected error.")

if __name__ == "__main__":
    calculator()
````

## Ternary Operator in Python

````python
# A Ternary Operator is a shorthand way of writing a simple if-else statement in a single line.

# It is called ternary because it involves three parts:

# Condition
# Value if condition is True
# Value if condition is False


#Syntax
#   value_if_true if condition else value_if_false
````

````python
a = 6
b= 6
print(True if a>=b  else False)
````

````python
age = 10

status = "Adult" if age>=18 else "Minor"
print(status)

# Normal if else
# age = 20

# if age >= 18:
#     status = "Adult"
# else:
#     status = "Minor"

# print(status)
````

````python
# Anatomy of Ternary Operator
# result = A if condition else B

# Where:

# condition → evaluated first
# If True → returns A
# If False → returns B
````

````python
x = 15

result = "Positive" if x>0 else "Negative"

print(result)
````

````python
## To check number is even or odd
check_number = int(input("Enter number to check even or odd"))

result = "Even" if check_number%2==0 else "Odd"
print(result)
````

````python
# Maximum of two number

a = 4
b = 6

max_num = a if a>b else b
print(f" Maximum number: {max_num}")
````

````python
# Positive/ Negative
x = -5
label = "Positive" if x >= 0 else "Negative"
print(label)
````

````python
## Ternary with function call

marks = 85
result = print("Pass") if marks >= 40 else print("FAil")
````

````python
# Ternary with expressions
a = 10
b = 20

result = (a + b) if a > 5 else (a - b)

print(result)
````

````python
## Nested ternary opertator

marks = 90
grade = (
    "A" if marks >= 90 else "B" if marks >= 80 else "C" if marks >= 70 else "Fail"
)

print(grade)

# Equivalent to if-elif-else
# if marks >= 90:
#     grade = "A"

# elif marks >= 80:
#     grade = "B"

# elif marks >= 70:
#     grade = "C"

else:
    grade = "Fail"
````

````python
# Ternary with List
numbers = [1, 2, 3, 4, 5]

result = [
    "Even" if n % 2 == 0 else "Odd"
    for n in numbers
]

print(result)
````

````python
# Ternary with tuple
number = tuple(numbers)

result = [ "Even" if num%2==0 else "Odd" for num in number]
print(result)
````
