# 1. What is a Set?

A **set** is an **unordered**, **mutable** collection of **unique elements**.

### Syntax

```python
numbers = {1, 2, 3}

languages = {"Python", "Java", "C++"}

empty_set = set()
```

❌ Incorrect

```python
empty = {}
```

This creates a **dictionary**, not a set.

---

# 2. Key Characteristics

| Property               | Description |
| ---------------------- | ----------- |
| Ordered                | ❌ No        |
| Indexed                | ❌ No        |
| Mutable                | ✅ Yes       |
| Duplicate Allowed      | ❌ No        |
| Heterogeneous          | ✅ Yes       |
| Hashable Elements Only | ✅ Yes       |

# 8. Mathematical Operations

These are extremely common.

---

## Union

All unique elements.

```python
A = {1,2,3}

B = {3,4,5}

A | B
```

Output

```python
{1,2,3,4,5}
```

Alternative

```python
A.union(B)
```

---

## Intersection

Common elements.

```python
A & B
```

Output

```python
{3}
```

Alternative

```python
A.intersection(B)
```

---

## Difference

Elements only in first set.

```python
A - B
```

Output

```python
{1,2}
```

Alternative

```python
A.difference(B)
```

---

## Symmetric Difference

Elements present in only one set.

```python
A ^ B
```

Output

```python
{1,2,4,5}
```

Alternative

```python
A.symmetric_difference(B)
```

# 14. Time Complexity

| Operation    | Complexity  |
| ------------ | ----------- |
| Add          | O(1)        |
| Remove       | O(1)        |
| Membership   | O(1)        |
| Union        | O(n)        |
| Intersection | O(min(n,m)) |
| Difference   | O(n)        |

Reason:
Python sets are implemented using **hash tables**, making lookups and insertions very fast on average.

# 16. Common Interview Questions

### Q1. Difference between List and Set?

| List                     | Set                      |
| ------------------------ | ------------------------ |
| Ordered                  | Unordered                |
| Allows duplicates        | Unique values only       |
| Indexed                  | No indexing              |
| Slower membership (O(n)) | Faster membership (O(1)) |

---

### Q2. Difference between remove() and discard()?

| remove()                   | discard()           |
| -------------------------- | ------------------- |
| Raises KeyError if missing | No error if missing |

---

### Q3. Difference between add() and update()?

| add()       | update()          |
| ----------- | ----------------- |
| One element | Multiple elements |

---

### Q4. Can a set contain a list?

No.

```python
{1,2,[3]}
```

Error

```
TypeError
```

Lists are mutable and therefore unhashable.

---

### Q5. Can a set contain tuples?

Yes.

```python
{
(1,2),
(3,4)
}
```

Tuples are immutable and hashable (if all their elements are hashable).

---

# 17. Best Practices

* Use sets when uniqueness matters.
* Prefer sets over lists for frequent membership checks.
* Use `discard()` instead of `remove()` when you're unsure an element exists.
* Convert lists to sets to remove duplicates quickly.
* Use set operations (`|`, `&`, `-`, `^`) instead of manual loops whenever possible.

````python
# Unordered: Set elements do not have an index, and their order is not guaranteed.

collection = {1, 2, 4, "hello", "world"}
print(collection) # Example output: {'world', 1, 2, 4, 'hello'} - order can change upon printing
````

````python
# Immutable Elements: While sets themselves are mutable (elements can be added or removed), 
# the elements within a set must be immutable (hashable). 
# This means integers, floats, strings, booleans, and tuples can be set elements. 
# Lists and dictionaries cannot be set elements because they are mutable.

valid_set = {1, "hello", True, (1, 2)} # Valid elements
print(valid_set)
# Output: {1, True, 'hello', (1, 2)} (Order may vary)

# Invalid element (list is mutable/unhashable)
# invalid_set = {1, [2, 3]} # This would raise TypeError: unhashable type: 'list'
````

Hashable vs. Unhashable: Hashable types have a consistent hash value that doesn't change during their lifetime, allowing them to be uniquely identified (essential for sets and dictionary keys). Mutable types like lists and dictionaries can change, making their hash values unstable and thus unhashable.

````python
# Empty Sets
# An empty set cannot be created with {} because that syntax creates an empty dictionary. To create an empty set, use # the set() constructor.

empty_dict = {}
print(type(empty_dict))
# Output: <class 'dict'>

empty_set = set()
print(type(empty_set))
# Output: <class 'set'>

empty_set.add(1)
print(empty_set)
print(type(empty_set))
# Output: {1}
````

````python
# Set Methods
# .add(element): Adds a new element to the set. If the element already exists, nothing happens (due to uniqueness).

collection = {1, 2}
collection.add(3)
collection.add(8787) # Duplicate, will be ignored
print(collection)
# Output: {1, 2, 3}
````

````python
# .remove(element): Removes a specified element from the set. If the element is not found, it raises a KeyError.

collection = {1, 2, 3, "Ravi"}
collection.remove("Ravi")
print(collection)
# Output: {2, 3}

# collection.remove(7) # This would raise a KeyError

collection.discard(5)    # error will not raise even if element is not present.
print(collection)
````

````python
# .clear(): Removes all elements from the set, making it empty.

collection = {1, 2, 3}
collection.clear()
print(collection) 
# Output: set()
print(len(collection))
# Output: 0
````

````python
# .pop(): Removes and returns a random element from the set. Since sets are unordered, there's no way to predict which element will be removed.

collection = {"hello", "apna college", "world", "coding", "python"}
# popped_element = collection.pop()
# print(f"Popped: {popped_element}")
print(collection.pop())
print(collection)
# Example Output:
# Popped: world
# {'python', 'coding', 'apna college', 'hello'} (element removed and order shifted)

````

````python
#.union(other_set): Returns a new set containing all unique elements from both the current set and other_set. It does not modify the original sets.

set1 = {1, 2, 3}
set2 = {2, 3, 4}
union_set = set1.union(set2)
print(union_set)
# Output: {1, 2, 3, 4}
# print(set1) # Original set1 remains unchanged: {1, 2, 3}
# print(set2) # Original set2 remains unchanged: {2, 3, 4}
````

````python

#.intersection(other_set): Returns a new set containing only the common unique elements between the current set and other_set. It does not modify the original sets.

set1 = {1, 2, 3}
set2 = {2, 3, 4}
intersection_set = set1.intersection(set2) 
print(intersection_set)
# Output: {2, 3} 
# print(set1) # Original set1 remains unchanged: {1, 2, 3}
# print(set2) # Original set2 remains unchanged: {2, 3, 4}
````
