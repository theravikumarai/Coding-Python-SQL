# 1. What is a Loop?

A **loop** is a control structure that repeatedly executes a block of code until a condition is met or until all items in a sequence are processed.

**Purpose:**

* Avoid writing repetitive code.
* Process large datasets efficiently.
* Automate repetitive tasks.

# 21. Best Practices

✅ Prefer `for` loops when iterating over collections.

✅ Use `while` only when the number of iterations is unknown.

✅ Use `enumerate()` instead of manual indexing.

✅ Use `zip()` to iterate over multiple sequences together.

✅ Avoid deep nested loops when possible; prefer vectorized operations with **NumPy**, **Pandas**, or **PyTorch** for better performance.

10. The pass Statement
The pass statement is a null operation in Python. It means "do nothing." It's used as a placeholder where a statement is syntactically required but you don't want any code to execute.

When pass is needed: Python expects an indented block after if, for, while, function definitions, etc. If the block is empty, it results in an IndentationError. pass fills this requirement.

````python
# Example of an empty loop causing an error
# for i in range(5):
#     # No code here
# print("Some useful work")
# This would raise an IndentationError

# Using pass to avoid error
for i in range(5):
    pass # This loop does nothing
print("Some useful work")
````

````python
# Purpose of pass:

# Placeholder: To indicate that code will be written later.
# Avoiding errors: Satisfies syntax requirements for empty code blocks.
# Conditional blocks: Can be used in if/else statements where one branch does nothing.
# Example of pass in an if-else block
i = 3
if i > 5:
    pass # Do nothing if i is greater than 5
else:
    print("i is not greater than 5")
````

````python
# A tiny video bundle: 2 frames, where each frame is 2 rows high and 2 pixels wide
video_batch = [
    [ ["Red", "Blue"],  ["Green", "Yellow"] ], # Frame 0
    [ ["White", "Black"], ["Grey", "Orange"] ] # Frame 1
]

# 1. Outer Loop: Steps through each frame in the video
for frame in video_batch:
    
    # 2. Middle Loop: Steps through each row in that frame
    for row in frame:
        
        # 3. Inner Loop: Steps through each pixel in that row
        for pixel in row:
            print(f"Analyzing color: {pixel}")
````
