## 1️⃣ Positional Arguments

📌 Passed **by order**

```python
def add(a, b):
    return a + b

add(2, 3)   # ✅
add(3, 2)   # Different result
```

🔹 Order **must match** parameters

---

## 2️⃣ Keyword Arguments

📌 Passed **by name**

```python
def greet(name, msg):
    print(msg, name)

greet(msg="Hi", name="Ravi")  # ✅
```

🔹 Order **does not matter**

---

## 3️⃣ Default Arguments

📌 Uses **default value** if not provided

```python

    return x ** y

power(4, 3)  # 64
```

🔹 Default parameters must come **after non-default**

## 🔥 Interview One-Liners

* `*args` → tuple of positional arguments
* `**kwargs` → dictionary of keyword arguments
* Keyword arguments override positional order
* Default arguments evaluated **once** (mutable danger ⚠️)

---

## ⚠️ Common Pitfall (Mutable Default)

```python
def add_item(item, lst=[]):  # ❌
    lst.append(item)
    return lst
```

✅ Fix:

```python
def add_item(item, lst=None):
    if lst is None:
        lst = []
    lst.append(item)
    return lst
```

---

## 📌 Ultra-Short Summary

```
Positional → order matters
Keyword → name matters
Default → fallback value
*args → many positional → tuple
**kwargs → many keyword → dict
```

````python
## Global Keyword: The global keyword is used when you want to modify a variable defined outside a function (a global variable) from inside a function.
count = 0

def increment():
    """ 
    global keyword --> "count" tells Python to look for the count variable that is outside the function.
    """
    global count  
    count += 1

print(count)
increment()
print(count)
````

````python
# 1. Training Step Counter

training_steps = 0
def train_batch():
    global training_steps
    training_steps += 1

train_batch()
train_batch()
print(training_steps)
````

````python
# Global Learning Rate

learning_rate = 0.01
def decay_lr():
    global learning_rate
    learning_rate *= 0.9

decay_lr()
print(learning_rate)
````

# Types of Functions in Python
    Built-in Functions These are functions that are pre-defined and readily available in Python. Examples:

        print(): For outputting values.
        len(): Returns the length of an object (string, list, tuple, etc.).
        type(): Returns the type of an object.
        range(): Generates a sequence of numbers.
        print() Function Parameters (sep and end) The print() function has optional parameters sep and end that control its output formatting.

        sep (separator): Specifies how to separate multiple arguments passed to print(). Default is a single space (' ').
        end (end character): Specifies what to print at the end of the output. Default is a newline character ('\n').
