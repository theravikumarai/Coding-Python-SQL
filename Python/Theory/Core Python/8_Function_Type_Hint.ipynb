# 1. Type Hints

## Definition

Type hints tell Python (and other developers) **what type of data is expected**.

They **do not stop** incorrect types at runtime.

```python
def add(a: int, b: int) -> int:
    return a + b
```

### Intuition

Think of type hints like **labels on containers**.

Without labels:

```
Box
??
```

With labels:

```
Box
Contains Integer
```

Anyone immediately knows what should go inside.

---

## Why Use Type Hints?

✅ Better readability

✅ IDE autocomplete

✅ Easier debugging

✅ Detect bugs before running

✅ Standard in production AI/ML code

# 7. Generic Types

## Problem

Suppose you write

```python
def first(items):
    return items[0]
```

What does it return?

Nobody knows.

---

## Solution

```python
from typing import TypeVar

T = TypeVar("T")

def first(items: list[T]) -> T:
    return items[0]
```

---

### Examples

```
Input

[1,2,3]

↓

Returns int
```

```
Input

["A","B"]

↓

Returns str
```

Same function.

Different types.

---

## Intuition

Generic = Universal Container

```
Bottle

↓

Can store

Water

Milk

Juice

Oil
```

Bottle doesn't care.

Same design.

Different content.

Generics work exactly like that.

---

# 8. Generic Class

```python
from typing import Generic, TypeVar

T = TypeVar("T")

class Box(Generic[T]):

    def __init__(self, item: T):
        self.item = item
```

Usage

```python
Box[int](10)

Box[str]("Hello")
```

Same class.

Different data types.

---

# 9. Protocol

Protocols describe **what an object can do**, instead of **what class it belongs to**.

---

Example

```python
from typing import Protocol

class Savable(Protocol):

    def save(self) -> None:
        ...
```

Any class having

```
save()
```

can be used.

---

### Intuition

Think about a USB port.

```
Keyboard

Mouse

Printer

Pendrive
```

Different devices.

Same USB interface.

Protocols work exactly like USB.

As long as the object has the required method,

it works.

# 10. typing Module

The **typing** module provides advanced type hints.

```python
from typing import (
    Any,
    Optional,
    Union,
    TypeVar,
    Generic,
    Protocol
)
```

Python 3.9+ also supports built-in generic syntax like:

```python
list[int]
dict[str, int]
tuple[int, str]
set[str]
```

instead of importing `List`, `Dict`, `Tuple`, or `Set`.

# 12. Best Practices

✔ Always type function parameters.

✔ Always type return values.

✔ Type public APIs.

✔ Type ML pipelines.

✔ Use `list[int]` instead of `List[int]` in modern Python.

✔ Use `Path` instead of raw strings for file paths.

✔ Use `Optional` (or `| None`) when values may be missing.

✔ Use `Protocol` for interfaces.

✔ Use `TypeVar` only when writing reusable generic code.

# Interview Questions

### Q1. Do type hints enforce types?

**Answer**

No.

Python remains dynamically typed.

Type hints are mainly for documentation, IDE support, and static analysis tools like `mypy` or `pyright`.

---

### Q2. Difference between Union and Optional?

```
Union

One of many types

int | str
```

```
Optional

A type OR None

str | None
```

---

### Q3. Why use Generics?

To write reusable code that preserves type information without duplicating logic.

---

### Q4. What is Protocol?

A way to specify required behavior (methods/attributes) without requiring inheritance.

---

### Q5. Why are type hints important in AI/ML?

* Improve readability of complex pipelines
* Catch errors early
* Better IDE assistance
* Easier collaboration on production code
* Cleaner APIs for data processing and model serving

````python
from typing import List, Dict, Set, Tuple, Optional, Any, Callable

def calculate_sum(a: int, b: int) -> int:
    """
    Calculate the sum of two integers.
    
    Args:
        a (int): First number.
        b (int): Second number

    Returns:
        int: Sum of a and b
    """
    return a + b


def main() -> None:
    total_sum = calculate_sum(5, 6)
    print(total_sum)

if __name__ == "__main__":
    main()
````

````python

def average_age(male: int, female: int) -> float:

    """
    Calculate average age of a couple.

    Args:
        male (int) : Age of the male.
        female (int) : Age of the female.

    Returns:
        float: average age of a couple.

    """

    return (male + female)/2

def main() -> None:
    try:
        male_age = int(input("Enter male age: "))
        female_age = int(input("Enter female age: "))

        if male_age < 0 or female_age < 0:
            raise ValueError
        
        avg_age = average_age(male_age, female_age)
        print(f"Average age of the couple is {avg_age}")

    except ValueError:
        print("Please enter valid integer ages")

if __name__ == "__main__":
    main()
````

````python

def palindrome_check(value: str) -> bool:
    """
    Check value is plaindrome or not.

    Args:
        value (str): Input string.

    Returns: 
        bool: True if palindrome, False otherwise.

    """
    return value == value[::-1]
    
def main() -> None:
  
    value = input("Enter a value: ")

        
    if palindrome_check(value):
        print("Given value is palindrome")
    else:
        print("Given value is not a palindrome") 


if __name__ == "__main__":
    main()

````

````python
# Write a function to print the elements of a list in a single line.
from typing import List
cities = ["Delhi", "Gurgaon", "Noida", "Pune", "Mumbai", "Chennai"]

def print_list(items: List[str]) -> None:
    """
    Printing list in a single line.

    Args:
        items (list): Input as a List.

    Returns:
        None
    """
    print(" ".join(items))    # join() is more pythonic - Cleaner, Faster, More readable.

 
def main() -> None:

    print_list(cities)

if __name__=="__main__":
    main()
````
