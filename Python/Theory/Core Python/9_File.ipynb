# File Input/Output (File I/O) 

---

# 1. What is File I/O?

**File I/O = File Input + File Output**

* **Input** → Read data from a file.
* **Output** → Write data to a file.

### AI/ML Intuition

Think of an AI project like a factory:

```
Dataset Files
      │
      ▼
Read into Python
      │
Preprocessing
      │
Model Training
      │
Predictions
      │
Save Results
```

Without File I/O, an AI model cannot:

* Read datasets
* Save trained models
* Store predictions
* Log experiments

# 3. File Modes

| Mode | Purpose      | Creates File? | Existing Data   |
| ---- | ------------ | ------------- | --------------- |
| `r`  | Read         | ❌             | Safe            |
| `w`  | Write        | ✅             | Overwritten     |
| `a`  | Append       | ✅             | Preserved       |
| `x`  | Create New   | ✅             | Error if exists |
| `rb` | Read Binary  | ❌             | Safe            |
| `wb` | Write Binary | ✅             | Overwritten     |
| `r+` | Read & Write | ❌             | Preserved       |

### Interview Tip

* `w` destroys old content.
* `a` keeps old content and adds new data.
* Use `rb`/`wb` for images, models, audio, etc.

---

# 4. Reading Files

## `read()`

Reads the **entire file**.

```python
with open("data.txt", "r") as f:
    data = f.read()
```

**Used for:**

* Small files
* Config files
* Prompt templates

---

## `read(n)`

Reads **n characters**.

```python
f.read(10)
```

Useful for partial reading.

---

## `readline()`

Reads **one line at a time**.

```python
line = f.readline()
```

Useful for:

* Large log files
* Streaming datasets

---

## `readlines()`

Reads all lines into a **list**.

```python
lines = f.readlines()
```

Output:

```python
['line1\n', 'line2\n']
```

---

# 5. Iterating Through a File (Best for Large Files)

```python
with open("reviews.txt") as f:
    for line in f:
        print(line)
```

### Why?

Instead of loading the entire file into memory, Python reads one line at a time.

**Industry Usage**

* NLP datasets
* Chat logs
* Large text corpora
* Server logs

---

# 6. Writing Files

### Write Mode (`w`)

```python
with open("result.txt", "w") as f:
    f.write("Accuracy = 95%")
```

**Important**

* Creates the file if it doesn't exist.
* Deletes previous content if it exists.

---

# 7. Appending Files

```python
with open("log.txt", "a") as f:
    f.write("Epoch 5 completed\n")
```

### Used for

* Training logs
* Experiment history
* Prediction history

---

# 8. Why Use `with open()`?

Instead of

```python
f = open("data.txt")
...
f.close()
```

Use

```python
with open("data.txt") as f:
    data = f.read()
```

### Advantages

✅ Automatically closes the file

✅ Prevents memory/resource leaks

✅ Safer if exceptions occur

**Industry Standard:** Always prefer `with open()`.

# 12. Pickle

Used to save any Python object.

Save

```python
pickle.dump(model,
            open("model.pkl","wb"))
```

Load

```python
model = pickle.load(
            open("model.pkl","rb"))
```

### Why?

Avoid retraining the model every time.

---

# 13. Joblib

Preferred for Scikit-Learn models.

```python
from joblib import dump, load

dump(model,"model.joblib")

model = load("model.joblib")
```

Advantages

* Faster
* Better for NumPy arrays
* Less memory usage

# 17. Interview Questions

### Q1. Difference between `read()` and `readline()`?

| `read()`             | `readline()`         |
| -------------------- | -------------------- |
| Reads entire file    | Reads one line       |
| Higher memory usage  | Lower memory usage   |
| Best for small files | Best for large files |

---

### Q2. Difference between `w` and `a`?

| `w`                 | `a`               |
| ------------------- | ----------------- |
| Overwrites file     | Appends to file   |
| Deletes old content | Keeps old content |

---

### Q3. Why use `with open()`?

* Automatically closes files
* Handles exceptions safely
* Prevents resource leaks
* Recommended industry practice

---

### Q4. Why use Joblib instead of Pickle?

* Optimized for Scikit-Learn
* Faster serialization/deserialization
* Better handling of large NumPy arrays

---

### Q5. Why save trained models?

Training can take minutes to days. Saving the model lets you reuse it for inference without retraining.

````python
#Problem: Write a function that replaces all occurrences of the word "Java" with "Python" in the practice.txt file.
def replace_java_with_python(file_path):
    with open(file_path, "r") as f:
        content = f.read()
    
    updated_content = content.replace("java", "Python")
    
    with open(file_path, "w") as f:
        f.write(updated_content)
    
    print(f"All occurrences of 'Java' have been replaced with 'Python' in '{file_path}'.") 
replace_java_with_python(".\\Data\\practice.txt")
````

````python
#Question 3: Search if a Word Exists in a File
def search_word(file_path, word):
    with open(file_path, "r") as f:
        content = f.read()
    
    if word in content:
        print(f"The word '{word}' exists in the file '{file_path}'.")
    else:
        print(f"The word '{word}' does not exist in the file '{file_path}'.")

search_word(".\\Data\\practice.txt", "Python")
````

````python
#Question 4: Find the Line Number Where a Word First Appears
def find_word_line_number(filename, word_to_search):
    line_number = 1
    with open(filename, "r") as f:
        while True:
            line_data = f.readline()
            if not line_data: # If line_data is an empty string, it's the end of the file
                break
            
            if word_to_search in line_data:
                print(f"The word '{word_to_search}' was found on line {line_number} in {filename}.")
                return line_number # Return the line number of the first occurrence
            
            line_number += 1
    
    print(f"The word '{word_to_search}' was NOT FOUND in {filename}.")
    return -1 # Return -1 if the word is not found

# Call the function
find_word_line_number(".\\Data\\practice.txt", "Python")
find_word_line_number(".\\Data\\practice.txt", "programming")
find_word_line_number(".\\Data\\practice.txt", "JavaScript") # Search for another word
````

````python
#Question 5: Count Even Numbers from Comma-Separated Data in a File

def count_even_numbers_in_file(file_path):
    with open(file_path, "w") as f:
        f.write("1,2,45,55,86,76,84,90,32,11,23,44,66,78,88,99")

    with open(file_path, "r") as f:
        data = f.read()
        numbers = data.split(",")
        
        even_count = 0
        for num in numbers:
            if int(num) % 2 == 0:
                even_count += 1

    print(f"Total even numbers in the file '{file_path}': {even_count}")

count_even_numbers_in_file(".\\Data\\numbers.txt") 
````

````python
#Mode for both read and write is "r+"
# mode allows you to read from and write to the same file without truncating it.
#Modes
# 'r+' - Read and Write mode
# 'w+' - Write and Read mode (truncates the file)
# 'a+' - Append and Read mode

#Examples:
with open('.\\Data\\example.txt', 'r+') as f:    # r+ only works if the file already exists
    f.write("This is an example of r+ mode.\n")
    f.write("Appending more text in r+ mode.\n")
    f.seek(0)
    content = f.read()
    print(content)


with open('.\\Data\\example_w.txt', 'w+') as f:  # w+ creates the file if it does not exist
    f.write("This is an example of w+ mode.\n")
    f.write("Appending more text in w+ mode.\n")
    f.seek(0)
    content = f.read()
    print(content)

with open('.\\Data\\example_a.txt', 'a+') as f:   # a+ creates the file if it does not exist
    f.write("This is an example of a+ mode.\n")
    f.write("Appending more text in a+ mode.\n")
    f.seek(0)
    content = f.read()
    print(content)

````
