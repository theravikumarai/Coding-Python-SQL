## What is a Lambda Function in Python?

A **lambda function** is a small anonymous (nameless) function that can have any number of inputs but only **one expression**.

### Syntax

```python
lambda arguments: expression
```

Equivalent normal function:

```python
def add(a, b):
    return a + b
```

Lambda version:

```python
add = lambda a, b: a + b

print(add(10, 20))
# 30
```

---

# Intuition

Think of lambda as a **"quick one-time-use function"**.

Imagine you need a screwdriver once to tighten a screw.

* Creating a normal function = buying a toolbox.
* Creating a lambda = borrowing a screwdriver for 10 seconds.

When the logic is very small, lambda avoids writing a full function.

---

# Why Lambda Exists?

Without lambda:

```python
def square(x):
    return x * x

numbers = [1, 2, 3, 4]
result = list(map(square, numbers))
```

With lambda:

```python
numbers = [1, 2, 3, 4]

result = list(map(lambda x: x * x, numbers))

print(result)
# [1, 4, 9, 16]
```

Cleaner when the function is used only once.

---

# Lambda vs Normal Function

| Normal Function                 | Lambda Function       |
| ------------------------------- | --------------------- |
| Uses `def`                      | Uses `lambda`         |
| Can contain multiple statements | Only one expression   |
| Has a name                      | Usually anonymous     |
| Reusable                        | Often one-time use    |
| Easier for complex logic        | Best for simple logic |

# Reduce


## ✅ What is `reduce()`?

`reduce()` is a **higher-order function** that **reduces a collection (list, tuple, etc.) to a single value** by **repeatedly applying a function**.

It is available in:

```python
from functools import reduce
```

---

## 🧠 Core Intuition (One Line)

> **`reduce()` keeps combining elements two at a time until only one result remains.**

Think of it as **folding** a list into a single value.

---

## 🔹 Basic Syntax

```python
reduce(function, iterable)
```

or with an initial value:

```python
reduce(function, iterable, initializer)
```

* `function` → takes **two arguments**
* `iterable` → list / tuple / sequence
* `initializer` → optional starting value

---

## 1️⃣ Simple Example: Sum of Numbers

```python
from functools import reduce

nums = [1, 2, 3, 4]

result = reduce(lambda a, b: a + b, nums)
print(result)
```

### 🟢 Output

```
10
```

---

## 🔍 How this works internally (MOST IMPORTANT)

Step by step:

```
Step 1: a=1, b=2 → 3
Step 2: a=3, b=3 → 6
Step 3: a=6, b=4 → 10
```

Final result → `10`

---

## 2️⃣ Reduce with Initial Value

```python
reduce(lambda a, b: a + b, nums, 10)
```

Process:

```
10 + 1 → 11
11 + 2 → 13
13 + 3 → 16
16 + 4 → 20
```

Output:

```
20
```

---

## 3️⃣ Product of Numbers

```python
reduce(lambda a, b: a * b, [1, 2, 3, 4])
```

Output:

```
24
```

---

## 4️⃣ Finding Maximum Element

```python
reduce(lambda a, b: a if a > b else b, [3, 7, 2, 9, 5])
```

Output:

```

```

---

## 5️⃣ Reduce with Strings (Concatenation)

```python

print(sentence)
```

Output:

```
Python is awesome
```

---

## 6️⃣ Real-Life Analogy 🧠

Imagine **rolling numbers into a snowball**:

* Start with first two → combine
* Take result → combine with next
* Keep going → one final snowball

That’s `reduce()`.

---

## 7️⃣ Reduce vs Loop (Comparison)

### Without reduce

```python

for x in nums:
    total += x
```

### With reduce

```python

```

⚠️ Less readable for beginners

---

✅ When:

## 🧠 Interview-Ready Definition ⭐

# Sorted

````python
### 🔹 **`map()` — Clear Explanation with Intuition**

---

## ✅ What is `map()`?

`map()` is a **higher-order function** that **applies a function to each element of an iterable** and returns a new iterable containing the results.

---

## 🧠 Core Intuition (One Line)

> **`map()` means: “Apply this operation to every item.”**

Think of it as an **assembly line**:

* Each item passes through the **same machine (function)**
* Comes out **transformed**

---

## 🔹 Basic Syntax

```python
map(function, iterable)
```

* `function` → what to apply
* `iterable` → list, tuple, range, etc.
* Result → **map object** (lazy iterator)

---

## 1️⃣ Simple Example: Square Numbers

```python
nums = [1, 2, 3, 4]

result = map(lambda x: x * x, nums)
print(list(result))
```

### 🟢 Output

```
[1, 4, 9, 16]
```

---

## 🔍 What happens internally?

```
x = 1 → 1*1 → 1
x = 2 → 2*2 → 4
x = 3 → 3*3 → 9
x = 4 → 4*4 → 16
```

Each element is **independently transformed**.

---

## 2️⃣ `map()` with Normal Function

```python
def double(x):
    return x * 2

nums = [1, 2, 3]

print(list(map(double, nums)))
```

Output:

```
[2, 4, 6]
```

---

## 3️⃣ `map()` with Multiple Iterables

```python
a = [1, 2, 3]
b = [4, 5, 6]

result = map(lambda x, y: x + y, a, b)
print(list(result))
```

Output:

```
[5, 7, 9]
```

👉 Stops when the **shortest iterable ends**.

---

## 4️⃣ Real-World Example: Convert Strings to Integers

```python
nums = ["1", "2", "3"]

print(list(map(int, nums)))
```

Output:

```
[1, 2, 3]
```

---

## 5️⃣ `map()` vs List Comprehension

### List comprehension

```python
[x * x for x in nums]
```

### map

```python
list(map(lambda x: x * x, nums))
```

| map()                | List Comprehension |
| -------------------- | ------------------ |
| Functional style     | Pythonic style     |
| Lazy (iterator)      | Eager              |
| Faster in some cases | More readable      |

---

## 6️⃣ Common Use with `filter()` and `reduce()`

```python
from functools import reduce

nums = [1, 2, 3, 4, 5]

result = reduce(
    lambda a, b: a + b,
    map(lambda x: x * x, filter(lambda x: x % 2 == 0, nums))
)

print(result)
```

Steps:

1. `filter()` → even numbers
2. `map()` → square them
3. `reduce()` → sum

---

## 7️⃣ When NOT to use `map()`

❌ Complex logic
❌ Multiple conditions
❌ Poor readability

Use loops or list comprehension.

---

## 🧠 Interview-Ready Definition ⭐

> `map()` applies a given function to each item of an iterable and returns an iterator of the results.

---

## 🔑 Final Takeaway (Remember This)

> **`map()` transforms data element-by-element.**

---

## 📌 One-Line Memory Trick

```
map() = transform every item
```

---

If you want next, I can:

* Compare **map vs comprehension vs loop**
* Show **map in ML preprocessing**
* Explain **lazy evaluation advantage**

Just tell me 😊

````

# List Comprenhension: Create List from another List.

````python
# Normal loop
squares = []

for x in range(5):
    squares.append(x * x)

# List comprehension
squares1 = [x * x for x in range(5)]
print(squares)
print(squares1)
````

````python
# 2️⃣ With Condition (Filtering)
evens = [x for x in range(10) if x % 2 == 0]
print(evens)     # 👉 Only numbers satisfying condition are included.
````

````python
# 3️⃣ Transform + Filter Together (Very Common)
even_squares = [x * x for x in range(10) if x % 2 == 0]
print(even_squares)
````

````python
# 4️⃣ Using if–else in List Comprehension
result = ["Even" if x % 2 == 0 else "Odd" for x in range(5)]
print(result)
# ⚠️ Note: if–else comes before for.
````

````python
# 5️⃣ List Comprehension with Strings
word = "python"
vowels = [i for i in word if i in "aeiou"]

print(vowels)
````

````python
# 6️⃣ Nested List Comprehension
matrix = [[1, 2], [3, 4], [5, 6]]

flat = [num for row in matrix for num in row]

print(flat)
````

````python
# 7️⃣ List Comprehension with Function
def square(x):
    return x * x

result = [square(x) for x in range(5)]
print(result)
````

````python
# 8️⃣ When NOT to Use List Comprehension ⚠️

# ❌ Very complex logic
# ❌ Multiple nested conditions
# ❌ Readability suffers

# Use normal loops instead.


# If your loop can be read in one line, list comprehension is perfect.
# If not, don’t force it.
````

# Dictionary Comprehension

### 🔹 **Dictionary Comprehension — Clear Explanation with Intuition**

---

## ✅ Definition

**Dictionary comprehension** is a **concise way to create dictionaries** from iterables using a **single line of code**, similar to list comprehension.

---

## 🧠 Core Intuition (One Line)

> **Dictionary comprehension = create key–value pairs using logic in one line.**

Think:

> “For each item, compute a **key** and a **value**.”

---

## 🔹 Syntax

```python
{key_expression: value_expression for item in iterable if condition}
```

* `key_expression` → dictionary key
* `value_expression` → dictionary value
* `condition` → optional filter

---

## 1️⃣ Basic Example

### Normal way

```python
squares = {}
for x in range(5):
    squares[x] = x * x
```

### Dictionary comprehension

```python
squares = {x: x * x for x in range(5)}
```

Output:

```
{0: 0, 1: 1, 2: 4, 3: 9, 4: 16}
```

---

## 2️⃣ With Condition (Filtering)

```python
even_squares = {x: x * x for x in range(10) if x % 2 == 0}
```

---

## 3️⃣ Transform Existing Dictionary

```python
prices = {"apple": 100, "banana": 40, "mango": 60}

discounted = {item: price * 0.9 for item, price in prices.items()}
```

Output:

```
{'apple': 90.0, 'banana': 36.0, 'mango': 54.0}
```

---

## 4️⃣ Swapping Keys and Values

```python
data = {"a": 1, "b": 2, "c": 3}
swapped = {v: k for k, v in data.items()}
```

Output:

```
{1: 'a', 2: 'b', 3: 'c'}
```

---

## 5️⃣ Using `if–else` in Dict Comprehension

```python
nums = {x: ("Even" if x % 2 == 0 else "Odd") for x in range(5)}
```

Output:

```
{0: 'Even', 1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even'}
```

---

## 6️⃣ Creating Dictionary from Two Lists

```python
keys = ["name", "age", "city"]
values = ["Ravi", 25, "Delhi"]

result = {k: v for k, v in zip(keys, values)}
```

---

## 7️⃣ Nested Dictionary Comprehension

```python
matrix = [[1, 2], [3, 4]]

matrix_dict = {(i, j): matrix[i][j]
               for i in range(len(matrix))
               for j in range(len(matrix[0]))}
```

---

## ⚠️ When NOT to Use Dict Comprehension

❌ Complex nested logic
❌ Hard-to-read expressions
❌ Multiple side effects

Use normal loops for clarity.

---

## 🧠 Interview-Ready Definition ⭐

> Dictionary comprehension is a concise way to construct dictionaries by iterating over an iterable and defining key–value pairs in a single expression.

---

## 🔑 Final Takeaway (Remember This)

> **Dict comprehension answers two questions:
> “What is the key?” and “What is the value?”**
