A generator is a special type of function that returns values one at a time using the yield keyword instead of return.

👉 It does not store all values in memory
👉 Values are produced lazily (on demand) when ever required

🔹 Why Use Generators?

Memory efficient
Faster for large data
Ideal for streams, logs, large files, ML pipelines

# Generator vs Normal function

| Feature   | Normal Function    | Generator                 |
| --------- | ------------------ | ------------------------- |
| Returns   | All values at once | One value at a time       |
| Keyword   | `return`           | `yield`                   |
| Memory    | High               | Low                       |
| Execution | Starts fresh       | Resumes from last `yield` |

```python
## Creating a Generator: Two Ways
##1. Generator Function
def my_generator():
    yield 1
    yield 2
    yield 3

g = my_generator()

print(next(g))  # 1
print(next(g))  # 2
print(next(g))  # 3
```

```python
# Normal function:

def add():
    return 10

# When called:

add()

# Function runs and finishes.
```

```python
## How yield works

def count():
    yield 1
    yield 2
    yield 3
# when called:
g = count()
#Nothing executes yet.
#executes only when.
next(g)
# it returns only value and take pauses
#next call
next(g)
## it resumes from where it stopped
## Notice function remember its state
```

```python
### Generator expression similar to list comprehension.
# nums = [x*x for x in range(1,10)]   # List Comprehension
# nums

nums = (x*x for x in range(5))
print(next(nums))
print(next(nums))
print(next(nums))
print(next(nums))
```

```python
#Real world examples: Reading log file

# BAD Approach: Loads entire file into meamory
# with open(log.txt) as f:
#     lines = f.readline()

#Better
def read_log():
    with open("log.txt") as f:
        for line in f:
            yield line

for log in read_log():
    print(log)

# Only one line is loaded at a time.
# Used heavily in production systems.
```

# Key Interview Definition

A generator is a special iterator that lazily produces values using yield, maintaining its execution state between calls and generating data on demand rather than storing all values in memory.

```python
## Read a file:
def read_file(file_path):
    with open(file_path, "r") as file:
        for line in file:
            yield line.strip()

file_path = "log.txt"

for line in read_file(file_path):
    print(line)
```

```python
def print_nunmber(num):
    for n in range(1, num+1):
        yield n


g = print_nunmber(5)
for i in g:
    print(i)
```

```python
# Generator in ML / Data Processing: Batch Processing
def batch_generator(data, batch_size):
    for i in range(0, len(data), batch_size):
        yield data[i:i+batch_size]

data = list(range(200))

for batch in batch_generator(data, 10):
    print(batch)
```
