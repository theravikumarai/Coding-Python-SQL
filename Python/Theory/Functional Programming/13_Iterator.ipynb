Remembers where it stopped

````python
# 9. Creating Custom Iterator:
class Counter:

    def __init__(self, limit):
        self.limit = limit
        self.current = 1

    def __iter__(self):
        return self
    
    def __next__(self):

        if self.current <= self.limit:
            value = self.current
            self.current += 1
            return value
        raise StopIteration
    
c = Counter(2)
for i in c:
    print(i)
````

# Short Answer

### Use an Iterator when:

* You need complete control over iteration.
* You want to create a custom iterable object using a class.
* You need to maintain complex state.
* You need methods other than iteration.

### Use a Generator when:

* You only want to produce values one at a time.
* You want less code.
* You are processing large datasets.
* You want lazy evaluation.

In practice, **90% of the time generators are preferred over custom iterators**.

# Example 1: Custom Iterator

```python
class Counter:

    def __init__(self, limit):
        self.limit = limit
        self.current = 1

    def __iter__(self):
        return self

    def __next__(self):

        if self.current <= self.limit:
            value = self.current
            self.current += 1
            return value

        raise StopIteration
```

Usage:

```python
for num in Counter(5):
    print(num)
```

Output:

```python
1
2
3
4
5
```

A lot of code.

---

# Same Thing Using Generator

```python
def counter(limit):

    current = 1

    while current <= limit:
        yield current
        current += 1
```

Usage:

```python
for num in counter(5):
    print(num)
```

Output:

```python
1
2
3
4
5
```

Much simpler.

# Interview Answer

If asked:

**"Generator vs Iterator?"**

You can answer:

> Every generator is an iterator, but not every iterator is a generator. Generators are created using `yield` and automatically handle iteration state, while iterators require implementing `__iter__()` and `__next__()` manually. Generators are preferred for lazy loading and large AI/ML datasets because they are simpler and more memory efficient.
