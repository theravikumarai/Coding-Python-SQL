# What is a Decorator?

A decorator is a function that takes another function as input and returns a modified function.

Visualization:

           Function
               |
               v
         Decorator
               |
               v
      Enhanced Function
      
Decorator Structure
def decorator(func):   func --> function passing as an argument

    def wrapper():
        print("Before")

        func()

        print("After")

    return wrapper

```python
#### Example 1
def decorator(func):
    text = "This is an example"
    def wrapper():
        print(text)
        print("Before function")

        func()         ## executing 

        print("After function")

    return wrapper


def hello():
    print("Hello")

greet = decorator(hello)   ## Passing hello function as an argument
greet()


# # What you wrote:
# def hello():
#     print("Hello")
# hello = decorator(hello)

# # The modern way:
# @decorator
# def hello():
#     print("Hello")

```

```python
# @ Syntax: Python provides shortcut syntax:

# instead: greet = decorator(hello)

#use:
@decorator
def hello():
    print("hello")

hello()
# It's a samething
```

# Why Decorators Are Needed in AI/ML

# Suppose you have:
train_model()
evaluate_model()
predict()
save_model()

# Need logging everywhere.
Without decorator:

def train_model():
    print("Start")
    ...
    print("End")

Repeated code everywhere.

Decorator solves this.

```python
# Types of decorator:
# Type 1: Basic Decorator:

def logger(func):

    def wrapper():
        print("Function Started")

        func()

        print("Function Finished")

    return wrapper

@logger
def train():
    print("Training")

train()
```

```python
### Type 2: Decorator with arguments

# Normal decorator faiks for:

# predict("email text")

# Need: *args, **kwrgs

def logger(func):

    def wrapper(*args, **kwrgs):

        print("Started")
        result = func(*args, **kwrgs)
        print("Finished")

        return result
    return wrapper

@logger
def predict(text):
    return f"Prediction for {text}"

print(predict("My email data"))
```

```python
## AI/ Example
def logger(func):

    def wrapper(*args, **kwrgs):

        print("Started")
        result = func(*args, **kwrgs)
        print("Finished")

        return result
    return wrapper

@logger      ### decorator function
def classify_email(text):
    return f"Classification for '{text}' is: Support Ticket"

print(classify_email("description"))

### Every pridiction automatically logged
```

```python
### Type 3: timing Decorator

# Measure execution time.

import time

def timer(func):

    def wrapper(*args, **kwrgs):
        start = time.time()
        # print(start)
        result = func(*args, **kwrgs)
        end = time.time()
        # print(end)

        print("Time:", round((end - start),2))

        return result
    
    return wrapper

@timer
def train_model():
    time.sleep(3)
    print("Model Trained")

train_model()

# Used in:

# Model training
# Data preprocessing
# Feature engineering
```

```python
### Type 4: Logging Decorator
def log(func):
    def wrapper(*args, **kwrgs):
        print(f"Calling {func.__name__}")
        result = func(*args, **kwrgs)
        print(f"{func.__name__} completed")

        return result
    
    return wrapper

@log
def preprocess_data():
    pass

preprocess_data()
```

```python
### Type 5: Authentication Decorator
## Used in FastAPI

def authentication(func):
    def login(user):

        if user == "admin":
            return func(user)
        
        elif user == "raviikrds":
            return func(user)
        
        return "Access Denied"
    
    return login


@authentication
def dashboard(user):
    return "Welcome"

login_userid = input("Enter your login user ID: ")
print(dashboard(login_userid))


        
```

```python
### TYPE 6: DEcorators with parameters

# Decorator accepting its own arguments
"""
Structure:
decorator(arg)
Three nested functions required.
"""
def repeat(n):
    def decorator(func):
        def wrapper():

            for _ in range(n):
                func()

        return wrapper
    return decorator

@repeat(3)
def hello():
    print("Hello")
print(hello())
```

```python
# AI/ML example
def retry(n):
    def decorator(func):
        def wrapper():

            for _ in range(n):
                func()

        return wrapper
    return decorator

@retry(3)
def call_llm():
    pass

# If API fails, retry 3 times.
```

```python
import time
import random

def retry(max_attempts):
    def decorator(func):
        # FIX 1: Add *args and **kwargs so the function can accept text inputs
        def wrapper(*args, **kwargs):
            
            for attempt in range(1, max_attempts + 1):
                try:
                    # Try to run the actual LLM API call
                    return func(*args, **kwargs)
                    
                except Exception as error:
                    print(f"⚠️ Attempt {attempt} failed: {error}")
                    
                    if attempt == max_attempts:
                        print("❌ All retry attempts failed!")
                        raise error  # Crash gracefully after last attempt
                    
                    # Wait 1 second before trying again (Cool down time)
                    time.sleep(1)
                    
        return wrapper
    return decorator

# --- Simulating a flaky LLM API Call ---
@retry(max_attempts=3)
def call_llm(prompt):
    print(f"Sending prompt: '{prompt}' to LLM...")
    
    # Simulate a 70% chance of a network failure
    if random.random() < 0.7:
        raise ConnectionError("API Server Timeout")
        
    return "AI Response: Hello! How can I help you today?"

# --- Test the code ---
try:
    response = call_llm("Explain deep learning")
    print(f"🎉 Success! Result: {response}")
except Exception:
    print("Main program handled the final failure.")
```

```python
## Type 7: CLass Decorator: A Class Decorator is a function that takes an entire Python class as an input, modifies or extends it, and returns the updated class
import datetime
from datetime import datetime

# Define the class decorator function
def add_metadata(cls):
    #Add new properties straight into the class blueprint
    cls.created_at = datetime.now().strftime("%Y-%m-%d")
    cls.author = "Ravi's AI Team"
    return cls  ## Always return the modified class


# --- Test the Code ---


# Apply it to a simple model class
@add_metadata
class SimpleModel:
    def __init__(self, name):
        self.name = name



# create an object from the decorated class
my_model = SimpleModel("LinearRegression")

# You can access the injected variables instantly!
print(f"Model Name: {my_model.name}")
print(f"Created On: {my_model.created_at}")
print(f"Created By: {my_model.author}")
```

```python
def add_version(cls):
    cls.version = "1.0"

    return cls


@add_version
class Model:
    pass

print(Model.version)
```

```python
### Type 8: Built-In Decorators

@staticmethod
# No access to instance

class Math:

    @staticmethod
    def add(a,b):
        return a+b
    
```

```python
@classmethod   ## Access Class
class Model:
    count = 0
    def get_count(cls):
        return cls.count
    
```

```python
# @property ### Method makes behave like attribute
class Model:
    
    @property
    def version(self):
        return "1.0"
    

m = Model()

print(m.version)
```

## Real AI/ML Decorator Examples

```python
## Model training logger
import time

def training_logger(func):

    def wrapper(*args, **kwargs):

        print("Training Started")

        start = time.time()

        result = func(*args, **kwargs)

        end = time.time()

        print("Training Completed")

        print("Execution Time:", end-start)

        return result

    return wrapper



@training_logger
def train_model():
    time.sleep(2)
    print("Random Forest Training")

train_model()

```
