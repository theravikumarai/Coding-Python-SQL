# Nested Functions & Closures 

# 1. Nested Function

## Definition

A **nested function** is a function defined **inside another function**.

```python
def outer():

    def inner():
        print("Hello")

    inner()

outer()
```

---

## Structure

```text
Outer Function
│
├── Variable(s)
├── Inner Function(s)
└── Calls Inner Function
```

Example:

```text
train_pipeline()
│
├── load_data()
├── preprocess()
├── train()
└── evaluate()
```

---

# 2. Why Do We Use Nested Functions?

### 1. Hide Helper Functions

If a function is useful only inside another function, keep it nested.

✅ Good

```python
def train():

    def preprocess():
        pass
```

❌ Bad

```python
def preprocess():
    pass

def train():
    preprocess()
```

when `preprocess()` is never used elsewhere.

---

### 2. Organize Large Problems

Instead of writing one huge function:

```python
def train_model():
    ...
    ...
    ...
```

Break it into logical steps.

```python
def train_model():

    def load():
        pass

    def preprocess():
        pass

    def train():
        pass
```

---

### 3. Access Outer Variables

Nested functions can directly use variables from the outer function.

```python
def outer():

    x = 10

    def inner():
        print(x)

    inner()
```

Output

```text
10
```

# 5. Lexical Scope

```python

    x = 20

        print(x)

```

Output

```text

```

```text
Local
↓

Enclosing (Outer Function)
↓

Global
↓

Built-in
```

# 6. Closure

## Definition

A **closure** is a function that remembers variables from its outer function even after the outer function has finished executing.

---

## Example

```python
def create_loss_function(weight):

    def loss(y_true, y_pred):
        return weight * abs(y_true - y_pred)

    return loss
```

---

## Execution

```python
weighted_loss = create_loss_function(5)
```

Flow

```text
create_loss_function(5)

        │
        ▼

weight = 5

        │
        ▼

Create loss()

        │
        ▼

loss remembers weight

        │
        ▼

Return loss()
```

---

```python

```

means

```python

```

Memory

```text

y_pred = 7

weight = 5
```

Calculation

```text

↓

↓

↓

```

Output

```text

```

# 7. Why Does the Inner Function Remember Variables?

Normally:

```text
Function Ends

↓

Memory Destroyed
```

But if an inner function still needs a variable:

Python keeps it alive.

```text
Outer Function

weight = 5

↓

Function Ends

↓

Closure Stores

weight = 5
```

This stored environment is called a **Closure**.

````python
#AI/ML Example 1: Data Preprocessing (Scaling Features)
# In Machine Learning, you often need to scale data so that all features are between 0 and 1. 
# A nested function is perfect for keeping your scaling logic neat and clean

def create_scalar(data_min, data_max):
    # outer fun sets the limits based on  training data
    denom = data_max - data_min

    def scale(value):
        # nested function does the actual math
        return (value -data_min)/denom
    return scale

# imagine our data goes from 0 to 100
my_scalar = create_scalar(0, 100)

# Now we can safely scale any new data point
print(my_scalar(100))
````

````python
# Custom loss function in Deep Learning.
# When building Neural Networks (using tools like TensorFlow or PyTorch), 
# you sometimes need a custom loss function that requires extra settings, 
# like a penalty weight (\(\alpha \)).Since deep learning frameworks 
# expect loss functions to only accept y_true and y_pred, you use a 
# nested function to pass that extra weight.

def custom_loss(alpha):
    # outer function takes the extra hyperparameters

    def loss(y_true, y_pred):
        # Nested fun takes what deep learning Library expects
        error = (y_true - y_pred) ** 2
        penalty = alpha * abs(y_pred)
        return error + penalty
    return loss #Return the function to the model compiler

alpha = 0.01

# --- How to use it ---
# We make a loss function where the penalty weight is 0.1
my_total_loss = custom_loss(alpha)

# The model can now use 'my_total_loss' during training
print(my_total_loss(y_true=10, y_pred=8))  # Calculates error + penalty
````

# Step by Step Process to create a Nested Functions

````python

# Step 1: Create the Main Task: Suppose we want to train the model.

def train_pipeline():
    pass

# Step 2: Identify the helper Tasks:
# Training requires:

# loading data
# preprocessing
# training

# These helpers don't need to exist globally.

# Step 3: Define Helper Functions Inside Main function
def train_pipeline():

    def load_data():
        pass

    def preprocess():
        pass

    def train():
        pass

# Step 4: Call them
def train_pipeline():

    def load_data():
        print("Loading data")

    def preprocess():
        print("Preprocessing")
    
    def train():
        print("Training model")

    load_data()
    preprocess()
    train()

train_pipeline()

````

````python
# Data Preprocessing Pipeline without nested
# def remove_nulls(data):
#     pass
# def scale_data(data):
#     pass
# def preprocess(data):
#     pass

## With nested functions:
def preprocess_pipeline(data):

    def remove_nulls(data):
        return [x for x in data if x is not None]
    
    def scale_data(data):
        max_val = max(data)
        return [x/max_val for x in data]
    
    def amplify_data(data):
        return [x*x for x in data]
    
    clean = remove_nulls(data)
    scaled = scale_data(clean)
    amplify = amplify_data(scaled)

    return amplify

data = [10, 20, None, 40]

result = preprocess_pipeline(data)
print(result)


# Intuition
# Preprocessing Pipeline
# │
# ├── remove_nulls()
# └── scale_data()

# These functions are only useful inside preprocessing.


````

````python
# Training Pipeline
def train_model_pipeline():
    learning_rate = 0.01

    def initialize_weights():
        print("Weight initialized")

    def optimizer_step():
        print(f"Updating weight using LR ={learning_rate}")

    initialize_weights()
    optimizer_step()

train_model_pipeline()


# Notice:

# optimizer_step()
# can directly access:
# learning_rate
# from the outer function.
# This is one of the biggest reasons nested functions are used.
````

````python
### Example of closure 

## exponent
def power_factory(exponent):
    print(1)
    def raise_to_power(base):
        print(3)
        return base**exponent
    print(2)
    return raise_to_power  #this sends the inner function

# 1. Save the inner function to a variable
# This creates a function that raises numbers to the power.
print(4)
raise_to_power_5 = power_factory(5)

## Use your new function
#This posses 3 as the 'base'
result = raise_to_power_5(3)

print(result)     # Output: 243 (which is 3* 3* 3* 3* 3)
````

````python
## Learning Rate Schedular:

# In DL learning rate often changes during training
def learning_rate_schedular(initial_lr):
    lr = initial_lr
    print(lr)

    def get_lr(epoch):
        print(epoch)
        return lr * (0.95 ** epoch)

    return get_lr

schedular = learning_rate_schedular(0.01)
print((schedular(1)))
print((schedular(50)))

# Intuition
# Think of the closure as a trainer that remembers:

# initial_lr = 0.01
# without needing to pass it every time.
````

````python
## Prediction Threshold Function

def threshold_classifier(threshold):
    desc = "This is a spam detector"
    def predict(probability):
        print(desc)
        return 1 if probability >= threshold else 0
    
    #return predict, you are telling to Python: "Do not run this spam detector yet. 
    # Just package it up with its memory of the threshold and hand the whole machine back to me.
    return predict

spam_detector = threshold_classifier(0.7)# 1. You build a detector tool set at 70% threshold
result = spam_detector(0.85)
print(result)

# print(spam_detector(0.97))
# print(spam_detector(0.67))

# Intuition

# Different models can use different thresholds.

# spam_detector = threshold_classifier(0.7)
# fraud_detector = threshold_classifier(0.9)

# Each closure remembers its own threshold.

"""
2. What happens with return predict() (Incorrect)If you add the parentheses (), 
you are telling Python to run the inner function immediately right then and there 
inside the outer function.This causes two major problems:

Problem A (Missing Data): The predict function requires you to pass in a 
probability value (like predict(0.85)). If you just type return predict(), 
Python tries to run it with absolutely no data, and your code crashes 
instantly with a TypeError: missing 1 required positional argument.

Problem B (Wrong Output): Even if you did force a number in there, 
it would instantly calculate a final answer (like 0 or 1) and return 
just that number. You would completely lose the reusable function tool.
"""
````

````python
### Feature Normalizer: A common preprocessing task

def create_normalizer(mean, std):
    desc = "This is a age normalizer model"
    def normalize(x):
        print(desc)
        return (x - mean) / std

    return normalize

age_normalizer = create_normalizer(35, 10)
normalizer = age_normalizer(55)
print(normalizer)

# Intuition

# During inference, we must use the same training mean and standard deviation.

# The closure remembers:

# mean = 35
# std = 10
````

````python
### Model prediction counter:

def prediction_counter():

    count = 0

    def predict():
        nonlocal count

        count += 1

        print(f"Prediction #{count}")

    return predict

model_api = prediction_counter()
model_api()
model_api()


# Intuition: Keeps track of how many predictions the model has served.
````

````python
## Mini ML Model Closure:
## Suppose we train a simple linear regression model
## Formula: y = wx + b
# insted of defining a class, we can use closure.

def train_linear_model(weight, bias):
    model = "LinearRegression"
    def predict(x):   ### here x is the single input feature for pred
        print(f"Predicting {model}")
        return weight * x + bias
    
    return predict

model = train_linear_model(weight=2.5, bias=10)

prediction = model(5)  # input value
print(prediction)

prediction = model(8)
print(prediction)


# What Happens Internally?
# model = train_linear_model(2.5, 10)

# The closure stores:

# weight = 2.5
# bias = 10

# Later:

# model(5)

# becomes

# 2.5 * 5 + 10

````

````python
## Closure with nonlocal

# Suppose we want to update a model threshold dynamically

def threshold_manager(initial_threshold):
    threshold = initial_threshold

    def update(new_threshold):
        nonlocal threshold           # modify a variable that belongs to the outer function
        threshold = new_threshold

    def predict(prob):
        return prob >= threshold

    return update, predict

# Initialize the closure with a 0.5 threshold
update, predict = threshold_manager(0.5)

# FIX: Save the result to 'prediction' before printing it
prediction = predict(0.6)
print(prediction)  # Output: True (0.6 is greater than or equal to 0.5)

# Change the threshold to 0.9 using the nonlocal updater
update(0.9)

# Check the same 0.6 probability against the new threshold
prediction = predict(0.6)
print(prediction)  # Output: False (0.6 is now less than 0.9)

# nonlocal threshold
# allows modification of the outer variable.
````

## Closure vs Class

# Closure
    def create_model(w, b):

        def predict(x):
            return w*x + b

        return predict

# Class
    class Model:

        def __init__(self, w, b):
            self.w = w
            self.b = b

        def predict(self, x):
            return self.w*x + self.b
    
# Rule of Thumb
        Use Closure when you need 1–2 functions with small state.
        Use Class when you need many methods and complex state.
        In AI/ML, closures are commonly used for loss functions, callbacks, preprocessing functions, and schedulers, while classes are used for full models and pipelines.

# Interview Definition:
        A closure is a function object that remembers and can access variables from its enclosing scope even after the outer function has finished execution. It is commonly used in AI/ML to create configurable preprocessing functions, loss functions, schedulers, and prediction utilities without using classes.

````python
# Closure( Very Important)

def create_loss_function(weight):

    def loss(y_true, y_pred):
        return weight * abs(y_true - y_pred)

    return loss

weighted_loss = create_loss_function(5)

weighted_loss(10,8)
# print(weighted_loss(10,7))

# How?

# create_loss_function(5)
#             │
#             └── loss remembers weight = 5

# Even after the outer function finishes, the inner function remembers the variable.

# This is called a Closure.
````

````python
## Custom Activation Function:

def create_activation(alpha):

    def leaky_relu(x):
        if x > 0:
            return x
        return alpha * x
    return leaky_relu

activation = create_activation(0.001)

print(activation(-10))

# This is similar to how configurable components are built in deep learning systems.
````

````python
## Real AI Example Mini Training Loop
def train_network(data):

    learning_rate = 0.001

    def forward(x):
        return x * 2
    
    def loss(pred, target):
        return abs(pred - target)
    
    def update():
        print(f"Updating with LR={learning_rate}")

    for x,y in data:
        pred = forward(x)
        l = loss(pred,y)

        print("Loss", l)

    update()

data = [(2,5), (3,6)]

train_network(data)
    

# Structure:
# train_network()
# │
# ├── forward()
# ├── loss()
# └── update()
# All helper functions stay private to the training process.
````
