### Recursion

    Recursion is a programming concept where a function calls itself repeatedly until a specific condition (the base case) is met. It's an alternative way to achieve repetition, similar to loops, but can simplify code for certain complex problems. Any task solvable with loops can be solved with recursion, and vice-versa.

Key Components of Recursion:

    Work/Action: The task the function performs in each call.

    Recursive Call: The function calls itself, usually with a modified input that moves towards the base case.
    
    Base Case: A condition that stops the recursion. Without it, recursion would run infinitely, leading to a RecursionError (maximum recursion depth exceeded).

Recursion is one of those ideas that feels confusing at first because a function appears to call itself. 
The intuition becomes much clearer when you think of it as breaking a big problem into smaller versions of the same problem.

        Intuition: "Ask a Smaller You"

        Imagine you need to climb 100 stairs.

        Instead of thinking:

        "How do I climb 100 stairs?"

        Think:

        "How do I climb the last stair if I already know how to climb 99 stairs?"

        This creates a chain:

        Climb 100 stairs = climb 99 stairs + 1 step
        Climb 99 stairs = climb 98 stairs + 1 step
        ...
        Climb 1 stair = just take 1 step

        That last simple case is called the base case.

# Core components of Recursion

1. Base Case
        The simplest version of the problems

        if n == 0:
            return 1

        W/o a base case; recursion never stops

2. Recursive Case:
        Reduce the problem into a smaller version.

        return n * factorial(n-1)



```python
# Printing Numbers from N down to 1 Recursively

def show(n):
    # Base Case: Stop recursion when n becomes 0
    if n == 0:
        return
    
    # Work
    print(n)

    # Recursive call: call show() for n-1
    show(n-1)

show(3)
```

```python
def show_with_end(n):
    if n == 0:
        return

    print(n)
    show_with_end(n - 1)
    print("End for", n) # This will execute when stack unwinds

show_with_end(3)
```

```python
# Print  1 to 3 using recursiom.

def print_num(n):
    if n == 4:     # Base Case
        return
    print(n)   #work
    print_num(n+1)

    print(f"Stack Unwinding {n}")
print_num(1)

```

```python
## Print Numbers forward.

def print_n(n):
    if n==0:
        return
    print("forward",{n})
    print_n(n-1)
    print(n)

def print_rev(m):
    if m==0:
        return
    print(m)
    print_rev(m-1)
    print("unwinding",m)


# print_n(5)
print_rev(5)
```

```python
# Factorial
# n! = n * (n-1)!
# Base case:
# 0! = 1

def factorial(n):
    if n == 0:
        return 1

    return n * factorial(n-1)

factorial(5)

```

```python
# Sum of digits

def sum_digits(n):
    if n == 0:
        return 0
    
    return (n % 10) + sum_digits(n // 10)

sum_digits(12345)
```

```python
# Check Palindrome
def is_palindrome(s):
    if len(s) <= 1:
        return True
    return s[0] == s[-1] and is_palindrome(s[1:-1])

is_palindrome("racecar")
```

```python
# Binary search
def binary_search(arr, low, high, target):
    if low > high:    # base case
        return -1

    mid = (low + high) // 2

    if arr[mid] == target:
        return mid
    elif arr[mid] > target:
        return binary_search(arr, low, mid - 1, target)
    else:
        return binary_search(arr, mid + 1, high, target)
```

```python
# Recursion Optimization
def power(x, n):
    if n == 0:
        return 1
    half = power(x, n // 2)
    if n % 2 == 0:
        return half * half
    else:
        return x * half * half

```

```python
# Factorial

def factorial(n):
    if n == 0:
        return 1

    fact = n * factorial(n-1)
    return fact

fact = factorial(10)
print(fact)

# What is the complexity?

# Time: O(n)
# Space: O(n) (call stack)
```

```python
# Fibonacci Number: Find the nth fibonacci NUmber

# fib(n) = fib(n-1)+ fib(n-2)

def fib(n):
    if n <= 1:
        return n
    
    return fib(n-1) +fib(n-2)

# print(fib(34))

# This is inefficient
# time = o(2^n)


# Better
from functools import lru_cache

@lru_cache(None)
def fib1(n):
    if n <= 1:
        return n

    return fib1(n-1) + fib1(n-2)

print(fib1(34))
```

```python
## Faster with loop and zero memory
def fib_loop(n):
    if n <= 1:
        return n
    
    # Track just the last two numbers
    prev2, prev1 = 0, 1
    
    for _ in range(2, n + 1):
        # Calculate next number and slide the window forward
        current = prev1 + prev2
        prev2 = prev1
        prev1 = current
        
    return prev1

# Fast and uses almost zero memory
print(fib_loop(34))
print(fib_loop(1000))  # Works instantly without crashing!
```

```python
# Palindrome check
def palindrome(s):
    if len(s) <= 1:
        return True

    if s[0] != s[-1]:
        return False
    
    return palindrome(s[1:-1])
```
