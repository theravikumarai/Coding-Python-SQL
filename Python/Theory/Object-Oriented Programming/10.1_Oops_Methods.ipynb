## Object Oriented Programming


OOPs is a way of structuring code around objects into reusable and modular code structures, in the form of classes and objects. A class is a blueprint of the structure, and an object is an instance of a class that represents the data and behavior.



# Simple Definiton:
OOPs is a programming approach that organizes code into classes and objects, where a class defines the blueprint and objects represent real-world entities with data and behavior.

OOPs is a way of organizing code using classes and objects to model real-world entities with data and behavior.

A real-world entity is anything that has:
Properties (data)  → attributes
Actions (behavior) → methods

Example: Car -> Real world
Data: color, speed, fuel
Behavior: start, stop, accelerate

This concept is focus on using reuable code and implementing DRY principles.

`````python
class Car:                                        #Class

    def __init__(self, make, model):             # Methods
        self.makes = make                        # Instance
        self.model = model
    
    def get_make(self):
        return self.makes

    def get_model(self):
        return self.model

c1 = Car("Toyata", "Canary")
print(c1.get_make())
print(c1.get_model())
`````

`````python
# Sum of two objects

class Sum:

    def __init__(self, num1, num2):    # methods
        self.n = num1
        self.m = num2          # attribute

    def get_sum(self):
        result = self.n + self.m
        print(f"Sum of {self.n} and {self.m} is {result}")

sum = Sum(5,6)                 #Object
sum.get_sum()
print(sum.n)
`````

`````python
# Example  of Instance Attributes

class Subject:

    def __init__(self, sub_name):
        self.sub = sub_name
        self.fab_sub = None      # Instance Attributes

    def set_fav_sub(self, fav_subject):
        self.fab_sub = fav_subject
        print(f"My favourite subject is {self.fab_sub}")

    
    
s1 = Subject("Physics")
s2 = Subject("Maths")

s1.set_fav_sub("Machine_Learning")
s2.set_fav_sub("Deep_Learning")

print(s1.fab_sub)
print(s2.fab_sub)


s1.fab_sub = "AI"

print(s1.fab_sub)
print(s2.fab_sub)
`````

`````python
# Example  of Class Attributes

class Subject:

    sub_hobby = "Neural Network"     # class attributes

    def __init__(self, sub_name):
        self.sub = sub_name
        self.fab_sub = None

    def set_fav_sub(self, fav_subject):
        self.fav_sub = fav_subject
        print(f"My favourite subject is {self.fab_sub}")

    
    
s1 = Subject("Physics")
s2 = Subject("Maths")

s1.set_fav_sub("Machine_Learning")
s2.set_fav_sub("Deep_Learning")

print(s1.fav_sub)
s1.fav_sub = "AI"

print(s1.fav_sub)
print(s2.fav_sub)

print(s1.sub_hobby)
print(s2.sub_hobby)

Subject.sub_hobby = "Iron"     # modifying class attributes

print(s1.sub_hobby) 
print(s2.sub_hobby)
`````

# Methods: Special function associated with an object or class.

Types of Methods:

1. Instance Methods
2. Class Methods
3. Static Methods
4. Property Methods


`````python
# Instance Method: Purpose is to set or get details about instances. Here by default it passes self parameter as first parameter during object call.

class Car: 
    def __init__(self, make, model, year):       #Instance Method
        self.make = make
        self.model = model
        self.year = year

    def get_age(self):      #Instance Method
        current_year = 2026
        age = current_year - self.year
        return age

s1 = Car("Toyata", "canary", 2020)
age = s1.get_age()

print(age)

`````

# Class Method: Special method bound to class not the instance of a class.

A class method is a method that:

Belongs to the class, not to any specific object
Takes cls as the first parameter (instead of self)
Can access and modify class variables
Is defined using the @classmethod decorator

`````python
# Example

class Employee:
    Company = "Capgemini"
    Salary = 50000
    location = "Pune"

    @classmethod
    def merge_info(cls, first_name, last_name):
        cls.f = first_name
        cls.s = last_name
        print(cls.f + " " + cls.s)
        
    @classmethod
    def change_salary(cls,sal):
        cls.Salary = sal

Employee.merge_info("Ravi","Kumar")
Employee.Salary
Employee.change_salary(90000)
Employee.Salary

# Here we can create @classmethod to update class attributer.
#
`````

`````python
class Student:

    count = 0
    total_age = 0

    def __init__(self, name, age):       #Object creation phase
        self.name = name
        self.age = age
        Student.count += 1
        Student.total_age += age


    @classmethod
    def get_count(cls):             #it depends on all student, That means: No self, No object needed, Worked directly on class data
        return f"Total # of students {cls.count}"
    
    # Why not instance method?
    # Problems:
        # Needs an object even though it doesn’t use it
        # Conceptually wrong (method doesn’t belong to one student)
        # Bad design (interviewers dislike this)
    
    # Why not static method?
    # This will work, but problems
        # Hardcodes class name (Student)
        # Breaks inheritance
        # Not flexible
        
    @classmethod
    def get_average_age(cls):
        if cls.count == 0:
            return 0
        else:
            return f"{cls.total_age / cls.count}"


student1 = Student("Ravi Kumar", 27)  # Each time it creates new object, assign name and age to that object, increase Student.count, Student.total_age
student2 = Student("Anisha Kumari", 26)
student3 = Student("Rishav Kumar", 25)

print(Student.get_count())
print(Student.get_average_age())


#We use class methods here because the operations depend on class-level data shared by all instances,
# such as total count and average age. Instance methods would be conceptually incorrect, 
# and static methods would reduce flexibility and inheritance support.
    
`````

### Property Method: The property decorator is used to define
1. Getter = read value
2. Setter = change value (with rules)
3. Deleter = remove value

methods for a class attributes.

It allows you to define these methods as if they were 

`````python
# Example

class Student:

    def __init__(self, name, birth_year):
        self.name = name
        self.birth_year = birth_year

    @property
    def age(self):
        return 2026 - self.birth_year

s = Student("Ravi", 1998)
print(s.age)       #looks like an attribute, python secretly does s.age(), but you don't see the parenthesis.
# print(s.age())   #Wrong

# Without @property 
# def age(self):
#     return 2026 - self.birth_year

# We would need --> s.age()
# Less Clean, Less natural

# Why @property is important:
# 1. Protect data (Encapsulation)  -> s.age is allowed
# 2. Validation using setter
# @property allows controlled access to attributes by exposing methods as attributes.
`````

`````python
#Example

class BankAccount:
    def __init__(self, balance):
        self._balance = balance    #(_ means internal use)
    
    # Getter(Read Only Access)
    @property
    def balance(self):           #getter
        return self._balance
    
acc = BankAccount(10000000)
print(acc.balance)      # allowed with property

# Now suppose we want to set some values
acc.balance = 20000000    # Will get error: AttributeError: property 'balance' of 'BankAccount' object has no setter
print(acc.balance)

# to overcome from this error we use setter( Controlled update )


    

    
`````

`````python
# SETTER
# to overcome from this error we use setter( Controlled update )
class BankAccount:
    def __init__(self, balance):
        self._balance = balance    #(_ means internal use)
    
    # Getter(Read Only Access)
    @property
    def balance(self):           #getter
        return self._balance       # private instance

    @balance.setter       # @balance.setter are modifying the same property, not creating a new one. balance is a specific property instance.
    def balance(self, amount):          #setter
        if amount < 0:
            raise ValueError("Balance cannot be negative")
        self._balance = amount    #Value of balance will set


acc = BankAccount(10000000)
print(acc.balance)      # allowed with property

# Now suppose we want to set some values
acc.balance = 20000000    # Will get error: AttributeError: property 'balance' of 'BankAccount' object has no setter
print(acc.balance)
`````

`````python
#Deleter
class BankAccount:
    def __init__(self, balance):
        self._balance = balance    #(_ means internal use)
    
    # Getter(Read Only Access)
    @property
    def balance(self):           #getter
        return self._balance

    @balance.setter    
    def balance(self, amount):          #setter
        if amount < 0:
            raise ValueError("Balance cannot be negative")
        self._balance = amount    #Value of balance will set
    
    @balance.deleter
    def balance(self):               #deleter
        print("Deleting balance")
        # del self._balance
        self._balance = None #instead of deleting




acc = BankAccount(10000000)
print(acc.balance)      # allowed with property

# Now suppose we want to set some values
acc.balance = 20000000    # Will get error: AttributeError: property 'balance' of 'BankAccount' object has no setter
print(acc.balance)

del acc.balance        #delete balance
print(acc.balance)     # as _balance attributes gets deleted, so Error: AttributeError: 'BankAccount' object has no attribute '_balance'
# Deleter does NOT set value to None
# Deleter completely removes the attribute
`````

`````python
# Deletion is useful
# Sensitive data (passwords, tokens)
# Cleanup before object destruction
# Enforcing object lifecycle rules
`````

`````python
# Setter and deleter use the property name because they extend the same property object created by the getter.
# @property → creates one smart attribute

# @balance.setter → adds write control

# @balance.deleter → adds delete control

# All three belong to the same attribute
`````

`````python
#Getter
class Circle:
    def __init__(self, radius):
        self._radius = radius

    @property
    def radius(self):
        return self._radius       #private instance
    
c = Circle(5)
print(c.radius)
`````

`````python
# Setter & Deleter

class Person:
    def __init__(self, first_name):
        self.first_name = first_name
        self.last_name = ""

    @property
    def full_name(self):
        return f"{self.first_name} {self.last_name}".strip()

    @full_name.setter
    def full_name(self,name):
        parts = name.split()
        if len(parts) != 2:
            raise ValueError ("Full name must consist first name & last name")
        self.first_name =  parts[0]
        self.last_name = parts[1]
        

    @full_name.deleter          
    def full_name(self):
        del self.first_name
    
n = Person("Ravi")
print(n.full_name)
n.full_name = "Ravi Kumar"
print(n.full_name)
del n.full_name
print(n.full_name)
`````

# 1. What is OOP?

## Definition

Object-Oriented Programming (OOP) is a programming paradigm where we organize code using **objects**, which combine **data (attributes)** and **behavior (methods)** into a single unit.

Instead of writing separate functions for everything, we create objects that know **what they have** and **what they can do**.

---

## Real World Intuition

Imagine a **Car**

A car has:

- Brand
- Model
- Color
- Speed

These are **attributes (data).**

A car can:

- Start
- Stop
- Accelerate

These are **methods (behaviors).**

So,

```
Car
 ├── Data
 │     Brand
 │     Color
 │     Speed
 │
 └── Functions
       Start()
       Stop()
       Accelerate()
```

OOP models real-world entities exactly like this.

---

# Why OOP?

Without OOP

```
brand = "BMW"
speed = 100

start_car()
stop_car()
```

Data and functions are separate.

Hard to maintain.

---

With OOP

```python
car.start()
car.stop()
```

Everything belongs to the object.

Cleaner and easier to maintain.

---

# Advantages

- Code Reusability
- Better Organization
- Easy Maintenance
- Encapsulation
- Scalability
- Easy Testing

# 4. Attributes

## Definition

Attributes are variables that belong to an object or class.

They store data.

Python mainly has two types.

---

# Type 1: Instance Attribute

## Definition

Belongs to each object individually.

Every object has its own copy.

---

## Intuition

Every student has a different

- Name
- Age
- Roll Number

These should not be shared.

---

Example

```python
class Student:

    def __init__(self,name):
        self.name = name

s1 = Student("Ravi")
s2 = Student("Rahul")

print(s1.name)
print(s2.name)
```

Output

```
Ravi
Rahul
```

Each object stores different values.

---

Memory

```
Student

s1
 └── name = Ravi

s2
 └── name = Rahul
```

---

# Type 2: Class Attribute

## Definition

Shared among all objects.

Stored only once inside the class.

---

## Intuition

All students belong to the same college.

College name is common.

No need to store it inside every object.

---

Example

```python
class Student:

    college = "ABC University"

    def __init__(self,name):
        self.name = name

s1 = Student("Ravi")
s2 = Student("Rahul")

print(s1.college)
print(s2.college)
```

Output

```
ABC University
ABC University
```

---

Memory

```
Student

college = ABC University

       ↑
   shared by

s1
name = Ravi

s2
name = Rahul
```

---

Changing Class Attribute

```python
Student.college = "XYZ"

print(s1.college)
print(s2.college)
```

Output

```
XYZ
XYZ
```

---

# Instance vs Class Attribute

| Feature       | Instance Attribute    | Class Attribute |
|----------     |-------------------    |-----------------|
| Belongs to    | Object                | Class |
| Shared        | No                    | Yes |
| Created       | Inside __init__       | Inside Class |

# 5.4 Property Method

## Definition

Allows accessing a method like an attribute.

Instead of

```python

```

We use

```python

```

Suppose

```

```

```

# Property Decorators

Property methods support

- Getter
- Setter
- Deleter

---

# Getter

Returns value.

Example

```python
class Student:

    def __init__(self):
        self._age = 25

    @property
    def age(self):
        return self._age

s = Student()

print(s.age)
```

Output

```
25
```

---

# Setter

Updates value.

Decorator

```
@attribute.setter
```

Example

```python
class Student:

    def __init__(self):
        self._age = 25

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self,value):
        if value > 0:
            self._age = value

s = Student()

s.age = 30

print(s.age)
```

Output

```
30
```

---

## Why Setter?

Validation.

Example

```python
s.age = -10
```

Without setter

Invalid value stored.

With setter

Reject invalid input.

---

# Deleter

Deletes attribute.

Decorator

```
@attribute.deleter
```

Example

```python
class Student:

    def __init__(self):
        self._age = 25

    @property
    def age(self):
        return self._age

    @age.deleter
    def age(self):
        del self._age

s = Student()

del s.age
```

---

# Methods Comparison

| Method | First Parameter | Access Instance | Access Class | Decorator |
|----------|----------------|----------------|--------------|-----------|
| Instance | self           | ✅            | ✅             | None |
| Class     | cls           | ❌            | ✅            | @classmethod |
| Static | None             | ❌            | ❌            | @staticmethod |
| Property | self           | ✅            | Optional      | @property |

---

# When to Use Which?

| Situation                       | Method |
|------------                     |---------|
| Work with object data           | Instance Method |
| Modify shared class data        | Class Method |
| Utility/helper function         | Static Method |
| Computed attribute / Validation | Property Method |

# Interview Questions

## Q1. What is OOP?

**Answer**

Programming paradigm that combines data and methods into objects.

---

## Q2. Difference between Class and Object?

| Class | Object |
|--------|---------|
| Blueprint | Instance |
| No memory for object data | Occupies memory |

## Q3. What is self?

**Answer**

Reference to the current object.

---

## Q4. Difference between Instance and Class Attributes?

| Instance        | Class |
|-----------      |--------|

| Inside __init__ | Inside class |

---

## Q5. Difference between Instance, Class and Static Method?

| Instance    | Class      | Static |
|-----------  |--------    |--------|
| self        | cls        | No self/cls |
| Object Data | Class Data | Independent Utility |

---

## Q6. Can a Static Method access instance variables?

**Answer**

No.

It has no access to `self` or `cls` unless an object/class is explicitly passed as an argument.

---

## Q7. What is @property?

**Answer**

Converts a method into a read-only attribute-like interface, allowing method calls using attribute syntax.

---

## Q8. Why use Property instead of Getter Method?

Example

Instead of

```python
emp.get_salary()
```

Use

```python
emp.salary
```

Cleaner API and easier validation.

---

## Q9. Why use Setter?

**Answer**

To validate or control updates before assigning values.

Example

Reject negative age or invalid salary.

---

## Q10. Difference between Getter and Setter?

| Getter        | Setter |
|----------     |---------|
| Read data     | Update data |
| @property     | @property_name.setter |

---

## Q11. What does the underscore (`_age`) mean?

**Answer**

A single leading underscore is a **convention** indicating the attribute is intended for internal use (protected). It does **not** enforce access restrictions in Python.

# Key Takeaways

- A **Class** is a blueprint; an **Object** is its instance.
- **Instance Attributes** belong to individual objects.
- **Class Attributes** are shared across all objects.
- **Instance Methods** use `self` and work with object data.
- **Class Methods** use `cls` and work with class-level data.
- **Static Methods** are utility functions independent of object or class state.
- **Property Methods** expose computed or controlled values like attributes.
- Use **Getter**, **Setter**, and **Deleter** to encapsulate access, validation, and cleanup.
- Service-based interviews frequently ask differences between method types, attribute types, `self`, `cls`, and practical uses of `@property`.
