## Three types:
1. Public Member : Accessible anywhere from outside class.
2. Private Member: Accessible within the class.
3. Protected Member: Accessible within the class and its sub class.


`````python
class Employee:
    def __init__(self, name, salary):
        self.name = name                                 # Public
        self._project = project                          # Protected
        self.__salary = salary                           # Private
`````

`````python
# Private Member (__) r: Accessibe only within class, And we cannot access them directly from class object.

class Employee:
    def __init__(self, name, salary):
        self.name = name    # public data member
        self.__salary = salary

e = Employee("Ravi Kumar", 120000)
# print(e.__salary)   #Can't access private variable outside of its class: AttributeError: 'Employee' object has no attribute '__salary'


e.__salary = 45000      #But after assigning value to attribute we can access the private
print(e.__salary)
`````

`````python
# We can access the private members from outside  with two approaches

# 1. Public Method                          2. Name Mangling
`````

`````python
# 1. Public Methods:

class Employee:
    def __init__(self, name, salary):
        self.name = name    # public data member
        self.__salary = salary

    # Public instance method
    def show(self):
        print("Name: ", self.name, "\nSalary: ",self.__salary)


e = Employee("Ravi Kumar", 120000)
e.show()
`````

`````python
# 2. Name Mangling: We can directly access the private and protected variable
#    Name mangling can be created as an identifiers by adding two leading underscores and one trailing undersocres (_classname__datamemeber)

class Employee:
    def __init__(self, name, salary):
        self.name = name    # public data member
        self.__salary = salary



e = Employee("Ravi Kumar", 120000)

print("Name: ", e.name)
print("Salary: ", e._Employee__salary)      # Name Mangling

`````

`````python
# Protected Member: Accessible with in the class and sub calss.

# Prefix the member name with single underscore(_project)
# Note: Protected data member can be used when you implement Inheritance and want to allow data members access to only to child classes.

# Parent class
class Company:
    def __init__(self):
        self._project = "NLP"      ### Protected Member
        
# Child Class
class Employee(Company):
    def __init__(self, name):
        self.name = name
        Company.__init__(self)

    def show(self):
        print(f"Employee name: {self.name}")
        # Accessing protected member in child class.
        print(f"Working on project: {self._project}")

c = Employee("Ravi")      
c.show()   # Direct access to protected dataa member.
`````

###  Nothing is private in Python:
    We can access/modify the private member data in a controled way with getter and setter.

`````python
# Primary purpose of using getter and setter in OOPs is to ensure data encapsulation.

# Getter and Setter methods are often used when:
# 1. To avoid direct access to private variables.
# 2. To add validation logic for setting a value.

class Student:
    def __init__(self, name, age):
        self.name = name
        self.__age = age               #private member

    @property
    def get_age(self):                #Getter Method
        return self.__age
    
    @get_age.setter
    def set_age(self, age):           #Setter Method
        self.__age = age

a = Student("Ravi",28)

print('Name: ', a.name,'\nAge: ',a.get_age)

a.set_age = 31
print('Name: ', a.name,'\nAge: ',a.get_age)
`````

# 1. What is Encapsulation?

## Definition

**Encapsulation** is the process of **combining data (attributes)** and **methods (functions)** into a single unit (class) while **restricting direct access** to some data.

In simple words,

> **Encapsulation = Data + Methods + Controlled Access**

---

## Technical Definition

Encapsulation hides the internal implementation of an object and allows users to interact with it through a controlled interface.

---

# Real-World Intuition

Imagine an **ATM Machine**.

As a customer, you can:

- Insert Card
- Enter PIN
- Withdraw Money
- Check Balance

But you **cannot directly access**

- Bank Database
- Account Table
- Server Logic

Everything happens behind the scenes.

```
Customer

   ↓

+----------------+
|     ATM        |
|----------------|
| Withdraw()     |
| Deposit()      |
| Balance()      |
+----------------+

      ↓

Hidden Database
```

This is Encapsulation.

The user only sees **what is necessary**.

---

# Another Intuition

Think of a **Car**.

You only use

- Steering
- Brake
- Accelerator

You don't manually control

- Fuel Injection
- Engine Timing
- Gear Synchronization

These details are hidden.

---

# Why Encapsulation?

Without encapsulation

```python

```

Anyone can modify important data.

This may lead to:

- Invalid data
- Security issues
- Bugs
- Data corruption

---

With encapsulation

```python

```

The class decides whether the operation is valid.

# Benefits

- Protects data
- Hides implementation
- Allows validation
- Easier maintenance
- Better code organization

# Access Modifiers

Access modifiers define **how accessible variables and methods are**.

Unlike Java/C++, Python does **not enforce strict access control**.

Instead, it follows **naming conventions**.

Python has three access levels:

- Public
- Protected
- Private

---

# 1. Public Members

## Definition

Public members can be accessed from **anywhere**.

No underscore is used.

---

## Syntax

```python
class Student:

    def __init__(self):
        self.name = "Ravi"
```

---

Example

```python
class Student:

    def __init__(self):
        self.name = "Ravi"

s = Student()

print(s.name)
```

Output

```
Ravi
```

---

Memory

```
Student

name

Accessible Everywhere
```

---

## When to Use

Use for data that should be freely available.

Example

- Student Name
- Employee ID
- Product Name

---

# 2. Protected Members

## Definition

Protected members are intended to be used

- inside the class
- inside child classes

They are prefixed with a **single underscore**.

```
_variable
```

---

## Important

Python **does not prevent** external access.

It is only a convention saying:

> "Please don't access this directly."

---

Example

```python
class Student:

    def __init__(self):
        self._age = 25

s = Student()

print(s._age)
```

Output

```
25
```

Although accessible, developers understand it should not be used directly.

---

## Intuition

Imagine an office.

Employees can enter the staff room.

Visitors technically could enter if no one stops them, but the sign says

```
Staff Only
```

Protected members work similarly.

---

## When to Use

Internal variables.

Example

- Cached values
- Temporary data
- Helper methods

---

# 3. Private Members

## Definition

Private members are intended to be accessed only inside the class.

Use **double underscore**.

```
__variable
```

---

Example

```python
class Bank:

    def __init__(self):
        self.__balance = 10000

b = Bank()

print(b.__balance)
```

Output

```
AttributeError
```

---

## Why?

Python performs **Name Mangling**.

Internally,

```
__balance

becomes

_Bank__balance
```

So the attribute exists but under a different internal name.

---

Example

```python
class Bank:

    def __init__(self):
        self.__balance = 10000

b = Bank()

print(b._Bank__balance)
```

Output

```
10000
```

---

## Intuition

Think of a mobile phone.

Users cannot directly access the operating system files.

Only the OS knows how to access them.

---

# Name Mangling

Python changes

```
__balance
```

into

```
_Bank__balance
```

This mechanism is called **Name Mangling**.

---

## Why Name Mangling?

It helps avoid accidental access and prevents subclasses from unintentionally overriding private members.

---

## Important Interview Point

Private members are **not truly private**.

They are only renamed internally.

Python trusts developers rather than enforcing strict access restrictions.

---

# Getter and Setter

Instead of accessing private data directly,

use methods.

---

Example

```python
class Bank:

    def __init__(self):
        self.__balance = 1000

    def get_balance(self):
        return self.__balance

    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount

b = Bank()

print(b.get_balance())

b.set_balance(5000)

print(b.get_balance())
```

Output

```
1000
5000
```

---

# Better Approach (Using Property)

```python
class Bank:

    def __init__(self):
        self.__balance = 1000

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, amount):
        if amount >= 0:
            self.__balance = amount

b = Bank()

print(b.balance)

b.balance = 5000

print(b.balance)
```

Output

```
1000
5000
```

This is the Pythonic way of implementing encapsulation.

# Public vs Protected vs Private

| Feature                | Public    | Protected | Private |
|----------              |---------  |-----------|----------|
| Prefix                 | None      | _         | __ |
| Accessible Everywhere  | ✅        | Yes (by convention) | ❌ (directly) |
| Child Class Access     | ✅        | ✅       | ❌ (directly) |
| Name Mangling          | ❌        | ❌       | ✅ |
| Intended Use           | Public API | Internal Use | Sensitive Data |

---

# Best Practices

✅ Use **public** members for data meant to be exposed.

✅ Use **protected** members for internal implementation details.

✅ Use **private** members for sensitive data such as passwords, balances, or tokens.

✅ Prefer `@property` over manual getter/setter methods for cleaner code.

---

Python has strict private variables.

❌ False

## Myth 2

Protected variables cannot be accessed.

❌ False

---

## Myth 3

Encapsulation means hiding all variables.

❌ False

# Interview Questions

## Q1. What is Encapsulation?

**Answer**

Encapsulation is the process of bundling data and methods into a single class while restricting direct access to internal data.

---

## Q2. Why do we need Encapsulation?

**Answer**

- Data protection
- Validation
- Better maintenance
- Improved security
- Cleaner APIs

---

## Q3. What are the access modifiers in Python?

**Answer**

- Public (`variable`)
- Protected (`_variable`)
- Private (`__variable`)

---

## Q4. Does Python support private variables?

**Answer**

Not strictly.

Python uses **name mangling** to discourage direct access.

---

## Q5. What is Name Mangling?

**Answer**

Python internally renames private members.

Example

```
__balance

↓

_Bank__balance
```

---

## Q6. Difference between Protected and Private?

| Protected             | Private |
|------------           |----------|
| `_var`                | `__var` |
| Convention only       | Name mangling |
| Accessible externally | Not directly accessible |
| Used for internal members | Used for sensitive data |

---

## Q7. Difference between Data Hiding and Encapsulation?

| Encapsulation                     | Data Hiding |
|---------------                    |-------------|
| Bundles data and methods together | Restricts direct access to data |
| Achieved using classes            | Achieved using access modifiers |

> **Data Hiding is one benefit (or aspect) of Encapsulation.**

---

## Q8. Why use `@property` instead of getter/setter methods?

**Answer**

- Cleaner syntax (`obj.balance` vs `obj.get_balance()`)
- Supports validation
- Maintains encapsulation
- More Pythonic

---

## Q9. Can we access private variables outside the class?

**Answer**

Not directly.

They can still be accessed using the mangled name:

```python
obj._ClassName__variable
```

However, this should be avoided.

---

## Q10. Can private methods be inherited?

**Answer**

Private methods are inherited, but they cannot be accessed directly by subclasses because of name mangling.

---

# Key Takeaways

- **Encapsulation** combines data and methods while controlling access.
- Python uses **naming conventions**, not strict access control.
- **Public (`var`)** → Accessible everywhere.
- **Protected (`_var`)** → Intended for internal/class hierarchy use.
- **Private (`__var`)** → Uses name mangling to discourage direct access.
- **Name Mangling** changes `__variable` to `_ClassName__variable`.
- Use **`@property`** for Pythonic encapsulation instead of traditional getter/setter methods.
- Encapsulation improves **security, maintainability, readability, and data integrity**.
- Frequently asked interview topics include **access modifiers, name mangling, encapsulation vs data hiding, and property decorators**.
