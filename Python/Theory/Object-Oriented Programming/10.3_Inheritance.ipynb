# Super Method: (super().__init__())

    Super method is used to access the method of a 'parent class' in the 'child class'.

    --> Not required to specify parent class name to access methods.
    --> We can use super() in both single and multiple inheritance.
    --> Super() supports code reusability as there is no need to write the entire functions.

`````python
# Currently, attributes are created inside methods. Better approach:

class Employee:
    def __init__(self, company):
        self.company = company

class Name(Employee):
    def __init__(self, company, name):
        super().__init__(company)
        self.name = name

class Age(Employee):
    def __init__(self, company, age):
        super().__init__(company)
        self.age = age

# object intilization
a = Age("Capgemini", 27)
b = Name("Capgemini", "Ravi Kumar")

print(a.company)
# print(a.name)
print(b.company)
print(b.name)
print(a.age)
`````

`````python
## Hybrid Inheritance: When multiple types of inheritance comes together.

class Vehicle:
    def vehicle_info(self):
        print("Inside Vehicle class")

class Car(Vehicle):
    def car_info(self):
        print("Inside Car Class")
    
class Truck(Vehicle):
    def truck_info(self):
        print("Inside Truck class")

class SportsCar(Car, Vehicle):
    def sports_car_info(self):
        print("Inside Sports Car class")

# object initialization
a = SportsCar()

a.vehicle_info()
a.car_info()
a.sports_car_info()
# a.truck_info()       # AttributeError: 'SportsCar' object has no attribute 'truck_info'
`````

# Super() Method or Function: 

    ## Super Method is used to access the method of a parent class in the child class.

    ## super() returns a temporary object of the parent class that allows us to call a parent class method inside a child class methods.

`````python
## super() used in both single and multiple inheritance. It supports code reusability as there is no need to write the entire function.

class Company:
    def company_name(self):
        return "Google"
    
class Car:
    def car_info(self):
        return "Toyota Canary"

class Employee(Company, Car):
    def __init__(self, name):
        self.name = name 

    def info(self):                              
        c_name = super().company_name()  # Using super() calling superclass method.
        c_car = super().car_info()        # Using super() calling superclass method.
        print(f"{self.name} works at {c_name} and has {c_car} car")
        
emp = Employee("Ravi Kumar")
emp.info()
`````

# Method Overriding:

    ## When a child class method has the same name, same parameters and same return type as a method in its superclass(parent class), then the method in the child class said to override the method in the parent class

### super(). __init__()    ----> class constuctor of the class.


`````python
# Method Overridding example.
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def speak(self):
        print("The animal speaks")

class Cat(Animal):
    def __init__(self, name, age, color):
        super().__init__(name, age)
        self.color =color
        
    def speak(self):                                 # Method Overridden
        print("Meow")

cat = Cat("Tony",3,"Gray")
print(cat.age)
print(cat.name)
print(cat.color)

cat.speak()    # Here, This happens because of Method Overriding and runtime polymorphism in Python.
# Reason behind it.
# cat is an object of Cat, Python looks for speak() inside the Cat class first
# Cat overrides the speak() method of Animal
# So Python executes Cat.speak(), not Animal.speak() 

`````

# Method Resolution Order (MRO): 

    ## C3 linearization ensures that overridden methods are always found in child classes before parent classes, without ambiguity.
    Python follows this order:

            Object’s class (Cat)---->Parent class (Animal)---->object (base class)

            Since Cat has its own speak() method, Python stops searching there.

    ## A parent class must always come AFTER its child in multiple inheritance.
    ### C3 linearization is the algorithm Python uses to calculate MRO (Method Resolution Order) in multiple inheritance
    ### It converts a complex inheritance tree into one straight line (linear order) that Python can follow safely.


`````python
### MRO(Method Order Resolution)8
class A:
    def process(self):
        print("In class A")
    
class B(A):
    def process (self):
        print("In Class B")

# class C(A,B):     ### Left to Right direction,  Error:  This violates C3 linearization rules, so Python refuses to build MRO.\

class C(B,A):
    pass

c1 = C()
c1.process()

print(C.mro())      ### It shows the direction
`````

`````python
# Example of issubclass()

from typing import List

class Company:
    def fun1(self):
        print("Inside Company class")

class Employee(Company):
    def fun2(self):
        print("Inside Employee classs")

class Player:
    def fun3(self):
        print("Inside Player class")


print(issubclass(Employee, Company))   # True
print(issubclass(Employee, List))      #Flase  ---> Built in functiion not class
print(issubclass(Player, Company))     # Flase
print(issubclass(Employee,(List, Company)))  # True due to Company class
print(issubclass(Company, (List, Company)))   # True due to company
print(Employee.mro())
`````

# 1. What is Inheritance?

## Definition

**Inheritance** is an OOP concept where one class (Child/Derived class) acquires the properties and behaviors of another class (Parent/Base class).

Instead of rewriting existing code, we **reuse** it.

---

## Technical Definition

Inheritance allows a child class to inherit:

- Attributes
- Methods

from a parent class and also add or modify its own functionality.

---

# Real-World Intuition

Imagine a **Father** and **Son**.

Father has:

- House
- Car
- Land

Son automatically inherits them.

Additionally, the son may have:

- Laptop
- Mobile
- Job

```
Father

 └── Land

        ↓

Son

 ├── Car (Inherited)
```

One child inherits from multiple parents.

```

    ↓
```

```
Car
Jewelry
```

# Method Resolution Order (MRO)

## Definition

MRO defines the order in which Python searches for methods in an inheritance hierarchy.

---

## Why MRO?

Suppose

Two parent classes have the same method.

Which one should Python call?

MRO decides.

---

Example

```python
class A:

    def show(self):
        print("A")

class B:

    def show(self):
        print("B")

class C(A,B):
    pass

c = C()

c.show()
```

Output

```
A
```

Python searches

```
C

↓

A

↓

B
```

---

# View MRO

```python
print(C.__mro__)
```

or

```python
print(C.mro())
```

Output

```
(C, A, B, object)
```

---

## C3 Linearization

Python uses the **C3 Linearization Algorithm** to compute the MRO.

It ensures:

- Consistent order
- No ambiguity
- Each class appears only once

For interviews, remember:

> **Python uses C3 Linearization to determine MRO.**

You usually do **not** need to manually compute it in service-based interviews.

---

# Diamond Problem

```
      A
     / \
    B   C
     \ /
      D
```

Both B and C inherit from A.

If D calls a method,

Which path should Python follow?

MRO solves this.

---

Example

```python
class A:

    def show(self):
        print("A")

class B(A):
    pass

class C(A):
    pass

class D(B,C):
    pass

D().show()
```

Output

```
A
```

MRO

```
D

↓

B

↓

C

↓

A

↓

object
```

| Happens within the same class | Happens across inheritance |
