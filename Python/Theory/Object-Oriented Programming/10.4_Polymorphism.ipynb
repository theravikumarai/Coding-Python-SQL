# Polymorphism: 
    meaning "many forms," is the ability of an object to take on many forms.

    ## Common Example is operator overloading in Python. Where " + " has many form, With integer sum , with strings concatenate, with List --> List concatenation etc.

    ### Polymorphism is used to write flexible, reusable, and scalable code.
    
    ### It allows the same method call to behave differently for different objects, reducing coupling and improving maintainability.

### Polymorphism means: same action, different behavior depending on the object.

`````python
# Example

class Animal:
    def speak(self):
        pass               ## common Interface

class Dog(Animal):
    def speak(self):
        print("Bark")

class Cat(Animal):
    def speak(self):
        print("Meow")
    
animals = [Dog(), Cat()]

for animal in animals:
    animal.speak()

## Here We call the same method ---> speak(), And we don't check the type.
# Python decides at runtime:
# If Dog → Bark
# If Cat → Meow
# 👉 This decision at runtime = polymorphism

# ❌ Without Polymorphism (Bad Code)
for animal in animals:
    if type(animal) == Dog:
        print("Bark")
    elif type(animal) == Cat:
        print("Meow")


# 🚫 More code
# 🚫 Hard to maintain
# 🚫 Not scalable
`````

`````python
# ML example of Polymorphism:

class Model:
    def predict(self, x):
        pass

class LinearModel(Model):
    def predict(self, x):
        return "Linear prediction"
        
class TreeModel(Model):
    def predict(self, x):
        return "Tree prediction"

models = [LinearModel(), TreeModel()]
for model in models:
    print (model.predict(models))


# ✔ Same call → predict()
# ✔ Different logic inside
# ✔ No if-else

# Polymorphism allows the same method call to behave differently for different objects, enabling flexible, reusable, and clean code.
`````

`````python
# Polymorphism via Base Class (tf.keras.Model)
# All Keras models inherit from tf.keras.Model.
# Custom Models (Advanced but Real)

import tensorflow as tf
import numpy as np

X_train_dense = np.random.rand(1000, 784)
X_train_cnn = np.random.rand(1000, 28, 28, 1)
y_train = np.random.randint(0, 10, size=(1000,))


class MyDenseModel(tf.keras.Model):
    def __init__(self):
        super().__init__()
        self.d1 = tf.keras.layers.Dense(64, activation="relu")
        self.out = tf.keras.layers.Dense(10, activation="softmax")

    def call(self, x):
        return self.out(self.d1(x))


class MyCNNModel(tf.keras.Model):
    def __init__(self):
        super().__init__()
        self.conv = tf.keras.layers.Conv2D(32, (3,3), activation="relu")
        self.flat = tf.keras.layers.Flatten()
        self.out = tf.keras.layers.Dense(10, activation="softmax")

    def call(self, x):
        return self.out(self.flat(self.conv(x)))

def train_model(model, X, y, epochs=5, batch_size=32):
    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy", # Assumes integer labels
        metrics=["accuracy"]
    )
    
# SAME usage
model1 = MyDenseModel()
model2 = MyCNNModel()

train_model(model1, X_train_dense, y_train)
train_model(model2, X_train_cnn, y_train)


# ✔ Both override call()
# ✔ Same training code works
# ✔ Runtime polymorphism
`````

`````python
# TensorFlow uses polymorphism by defining a common model interface (compile, fit, predict) through tf.keras.Model.
# Different deep learning models override internal behavior while sharing the same external API, 
# allowing interchangeable models without changing training or inference code.

# In TensorFlow, we change the model — not the training code.
# That’s polymorphism.
`````

## Method Overloading: 
    Process of calling the same method with different parameters.

    Python doesnot support method overloading, Python considers only the latest defined method. Even if you overload the method, Python will raise TypeError.

`````python
class Geometry:

    def area(self, radius):
        return 3.14*radius*radius

    def area(self, l, b):
        return l*b
    
obj = Geometry()

print(obj.area(4))   # TypeError: Geometry.area() missing 1 required positional argument: 'b'

# As it executes the area(l,b)  ----> latest methods

## Here it throws error, becouse the below area() required 2 arguments. But we have sent only 1, so it throws error.
`````

## To overcome from above function:
    We have to write the method logic, so it executes inside the function depending on the parameter passes.

`````python


class Geometry:
    def area(self, a, b=0):        ## Using default argument
        if b == 0:
            print('Circle', 3.14*a*a)
        else:
            print("Rectangle", a*b)


obj = Geometry()
obj.area(4)
obj.area(4,5)
`````

`````python
# By using Positional Variable Arguments(*args)
class Geometry:
    def area(self, *args):
        if len(args) == 1:
            r = args[0]
            print("Circle", 3.14 * r * r)

        elif len(args) == 2:
            a, b = args
            print("Rectangle", a * b)

        else:
            print("Invalid number of arguments")

obj = Geometry()
obj.area(4)
obj.area(4,5)
`````

### Python implements this using special methods called dunder methods (double underscore methods). 
    
    For example, when you use + on two objects, Python internally calls the __add__ dunder method of the first object.

    Implementing Operator Overloading with Dunder Functions (Explicit): We can define these dunder methods in our custom classes to give operators specific behavior for our objects.

    Operators in python can be overloaded using dunder methods. And called when used in the objects.


`````python
# Examples
class Number:
    def __init__(self, num):
        self.num = num
    
    def __add__(self,other):
        print("Lets add")
        return Number(self.num + other.num)

n1 = Number(4)
n2 = Number(6)
n3 = Number(12)
result = n1 + n2 + n3
print(result.num)
`````

## Example: 
    Complex Number Class 
    
    Let's create a Complex number class and overload the + and - operators. Complex numbers have a real and an imaginary part (e.g., a + bj). When adding or subtracting, the real parts are added/subtracted, and the imaginary parts are added/subtracted separately.

`````python

class Complex:
    def __init__(self, real, imaginary):
        self.real = real
        self.imaginary = imaginary

    def show_number(self):
        print(f"{self.real}i + {self.imaginary}j")

    # Overload the '+' operator using the __add__ dunder method
    def __add__(self, other):
        new_real = self.real + other.real
        new_imaginary = self.imaginary + other.imaginary
        return Complex(new_real, new_imaginary) # Return a new Complex object

    # Overload the '-' operator using the __sub__ dunder method
    def __sub__(self, other):
        new_real = self.real - other.real
        new_imaginary = self.imaginary - other.imaginary
        return Complex(new_real, new_imaginary) # Return a new Complex object

# Create Complex number objects
num1 = Complex(1, 3)
num2 = Complex(4, 6)

print("Number 1:")
num1.show_number()
print("Number 2:")
num2.show_number()

# Use the overloaded '+' operator
num3 = num1 + num2 # Internally calls num1.__add__(num2)
print("Sum (num1 + num2):")
num3.show_number() # Expected: 5i + 9j

# Use the overloaded '-' operator
num4 = num1 - num2 # Internally calls num1.__sub__(num2)
print("Difference (num1 - num2):")
num4.show_number() # Expected: -3i + -3j
`````

`````python
class Num:

    def __init__(self, num):
        self.num = num

    def __add__(self, other):
        return self.num + other.num


n1 = Num(4)
n2 = Num(6)

result = n1 + n2
print(result)

# Flow
# n1 ---------> Num object { num: 4 }
# n2 ---------> Num object { num: 6 }

# n1 + n2
# ↓
# n1.__add__(n2)
# ↓
# self.num + other.num
# ↓
# 4 + 6
`````

`````python
class Num:

    def __init__(self, num):
        self.num = num

    def __sub__(self, other):
        return self.num - other.num


n1 = Num(4)
n2 = Num(6)

result = n1 - n2
print(result)
n1.__sub__(n2)
`````

- Special Methods

## Q3. What is Duck Typing?

**Answer**

Duck typing means Python determines compatibility based on whether an object provides the required methods or attributes, rather than checking its class.

---

## Q4. Does Python support Method Overloading?

**Answer**

Not in the traditional sense.

Python achieves similar behavior using:

- Default arguments
- `*args`
- `**kwargs`

---

## Q5. What is Method Overriding?

**Answer**

Method overriding occurs when a child class provides its own implementation of a method inherited from the parent class.

---

## Q6. What is Operator Overloading?

**Answer**

Operator overloading allows operators such as `+`, `-`, and `==` to work with user-defined objects by implementing corresponding magic methods.

---

## Q7. What are Magic (Dunder) Methods?

**Answer**

Magic methods are special methods (e.g., `__init__`, `__str__`, `__add__`) that define how objects interact with Python's built-in syntax, operators, and functions.

---

## Q8. Difference between Method Overloading and Method Overriding?

| Method Overloading | Method Overriding |
|--------------------|-------------------|
| Same method name with different parameters | Same method name and signature in parent and child |
| Not directly supported in Python | Fully supported |
| Achieved using default arguments or `*args` | Achieved through inheritance |

---

## Q9. Difference between `__str__()` and `__repr__()`?

| `__str__()` | `__repr__()` |
|--------------|--------------|
| User-friendly string representation | Developer/debug representation |
| Used by `print(obj)` | Used by `repr(obj)` and interactive interpreter |

# Key Takeaways

- **Polymorphism** means one interface with multiple implementations.
- Python supports polymorphism through **built-in functions, inheritance, functions, duck typing, and operator overloading**.
- **Method Overriding** provides runtime polymorphism.
- **Duck Typing** focuses on an object's behavior rather than its class.
- Python does **not** support traditional **method overloading**, but similar behavior can be achieved using **default arguments** and `*args`.
- **Operator Overloading** customizes operators for user-defined classes using **magic methods**.
- **Magic (Dunder) Methods** allow custom classes to integrate seamlessly with Python's syntax and built-in functions.
- Service-based interviews frequently ask about **duck typing, method overloading vs overriding, operator overloading, and common magic methods**.
