```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def welcome(self): # An instance method
        print(f"Welcome student {self.name}!")

    def get_marks(self): # Another instance method
        return self.marks

s1 = Student("Karan", 97)
s1.welcome()      # Output: Welcome student Karan!
print(s1.get_marks()) # Output: 97
s1.marks = 100   # Modifying the marks attribute
print(s1.get_marks()) # Output: 100|
print(s1.marks)
```

Problem: Create a Student class that takes name and marks (list of 3 subjects) as arguments in its constructor. Then, create a method to print the average of all subjects.

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
        
    def avg_marks(self):
        # Directly sum the list and divide by its length
        avg = sum(self.marks) / len(self.marks)
        return f"Average marks of {self.name} is {avg:.2f}"

s1 = Student("Ravi", [85, 90, 78])
s2 = Student("Ajay", [88, 76, 95])

# Now the print function properly outputs the returned string
print(s1.avg_marks())
print(s2.avg_marks())
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    
    def avg_marks(self):
        total = 0
        for mark in self.marks:
            total += mark
        avg = total / len(self.marks)
        print(f"Average marks of {self.name} is {avg:.2f}")
s1 = Student("Ravi", [85, 90, 78])
s2 = Student("Ajay", [88, 76, 95])
print(s1.avg_marks())
print(s2.avg_marks())
```

```python
### Static Methods in Python Classes
class MathOperations:
    @staticmethod
    def add(a, b):
        return a + b

    @staticmethod
    def multiply(a, b):
        return a * b

print(MathOperations.add(5, 10))        # Output: 15
print(MathOperations.multiply(5, 10))   # Output: 50
```

# @classmethos is used to create class inside method:

```python
## Class Methods in Python Classes
class Student:
    school_name = "ABC School"  # Class attribute

    def __init__(self, name, age):
        self.name = name          # Instance attribute
        self.age = age            # Instance attribute

    @classmethod
    def change_school_name(cls, new_name):
        cls.school_name = new_name

s1=Student("Ravi", 21)
s1.change_school_name("XYZ School")
print(s1.school_name)  # Output: XYZ School
```

```python
class Employee:
    count = 0

    def __init__(self):
        Employee.count += 1

    @classmethod
    def total_employees(cls):
        return cls.count

e1 = Employee()
e2 = Employee()

print(Employee.total_employees())
```

Classmethod keeps creation logic inside the class
We use @classmethod to define alternative constructors and encapsulate object-creation logic in a way that is safe for inheritance and future changes.”



__init__ → how object is initialized

@classmethod → how & from where object is created



```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod                           # Class method to create instance from string
    def from_string(cls, info_str):
        name, age = info_str.split('-')    # Splitting the string, Logic inside class method
        return cls(name, int(age))

s1 = Student.from_string("Ravi-21")
print(s1.name, s1.age)  # Output: Ravi 21
```

```python
# Problem: Create an Account class with two attributes: balance and account_number. 
# Create methods to debit (subtract amount), credit (add amount), and get_balance (return current balance).

class Account:
    def __init__(self, account_number, balance=0):
        self.account_number = account_number
        self.balance = balance

    def debit(self, amount):
        if amount > self.balance:
            print("Insufficient funds")
        else:
            self.balance -= amount
            print(f"Debited {amount}. New balance is {self.balance}")

    def credit(self, amount):
        self.balance += amount
        print(f"Credited {amount}. New balance is {self.balance}")
        
    def get_balance(self):
        return self.balance

    def _display_current_balance(self):
        print(f"Current balance is {self.get_balance()}")

acc = Account("123456789", 1000)
acc.debit(200)
acc.credit(500)
acc._display_current_balance()
acc.credit(150000)
```
